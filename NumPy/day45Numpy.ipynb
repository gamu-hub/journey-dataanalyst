Mini Project – Student Score Improvement

```python
import numpy as np
scores = np.array([
    [70,80,90],
    [60,75,85],
    [95,88,92]
])

bonus = np.array([5,3,2])

new_scores = scores + bonus

print("Original Scores")
print(scores)

print("\nBonus")
print(bonus)

print("\nNew Scores")
print(new_scores)
```
