Today's Mini Project

Create this array:

scores = np.array([
    [80,90,75],
    [88,92,79],
    [95,85,91]
])

Print:

Number of dimensions
Shape
Size
Data type
First row
Last column
The value 92

```python
import numpy as np
scores = np.array([
    [80,90, 75],
    [88, 92, 79],
    [95, 85, 91]
])


print("Number of dimension",scores.ndim)

print("Shape",scores.shape)

print("Size =", scores.size)

print("Datatype =", scores.dtype)

print("First row is", scores[0])

print("Last column is", scores[:,2])

print("The Value 92 is",scores[1,1])
```
