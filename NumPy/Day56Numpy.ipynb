# NumPy Indexing & Slicing
1️⃣ Basic Indexing

Remember that NumPy uses zero-based indexing.

```python
import numpy as np

sales = np.array([100, 200, 300, 400, 500])

print(sales[0])
print(sales[2])
print(sales[-1])
```

```python
# 2️⃣ Indexing 2D Arrays

# This is where it becomes really useful for data analysis.

import numpy as np

sales = np.array([
    [120, 150, 180, 200],
    [300, 280, 320, 350],
    [500, 450, 470, 520],
    [250, 270, 260, 290]
])

print(sales[0]) # first row
print(sales[1]) # second row
print(sales[1, 2]) # Get value at row 2, column 3
```

3️⃣ Selecting Columns

Want the first column?

```python
import numpy as np

sales = np.array([
    [120, 150, 180, 200],
    [300, 280, 320, 350],
    [500, 450, 470, 520],
    [250, 270, 260, 290]
])
print(sales[:, 0]) # first column
print(sales[:, 1]) # 2nd column

```

4️⃣ Slicing

Slicing lets you select a range.

```python
import numpy as np

sales = np.array([100, 200, 300, 400, 500])
print(sales[1:4]) # start at index 1, stop before index 4
```

# 2D slicing

```python
import numpy as np

sales = np.array([
    [120, 150, 180, 200],
    [300, 280, 320, 350],
    [500, 450, 470, 520],
    [250, 270, 260, 290]
])
print(sales[:, 1:3])
```

5️⃣ Boolean Indexing

You already practiced this yesterday.

```python
import numpy as np

sales = np.array([
    [120, 150, 180, 200],
    [300, 280, 320, 350],
    [500, 450, 470, 520],
    [250, 270, 260, 290]
])
print(sales[sales > 300])
```

6️⃣ Combining Conditions

```python
import numpy as np

sales = np.array([
    [120, 150, 180, 200],
    [300, 280, 320, 350],
    [500, 450, 470, 520],
    [250, 270, 260, 290]
])
sales[(sales > 250) & (sales < 500)]
print(sales)
```

7️⃣ Fancy Indexing

```python
sales = np.array([
    [120, 150, 180, 200],
    [300, 280, 320, 350],
    [500, 450, 470, 520],
    [250, 270, 260, 290]
])

#You want rows 0, 2 and 3.
print(sales[[0,2,3]])

#to select specific columns
print(sales[:, [0, 2]]) #Give me every row but only columns 0 and 2
```

```python
import numpy as np

sales = np.array([
    [120, 150, 180, 200],
    [300, 280, 320, 350],
    [500, 450, 470, 520],
    [250, 270, 260, 290]
])

print("Printing the third row")
print(sales[2])

print("\nPrinting the second column")
print(sales[:, 1])

print("\nPrinting the value 470")
print(sales[2,2])

print("\nPrinting the first two rows")
print(sales[0:2])

print("\nPrinting the last two columns")
print(sales[:, 2:4])

print("\nprinting all values greater than 300")
print(sales[sales > 300])

print("\nprinting only rows 0 and 3 using fancy indexing")
print(sales[[0,3]])

print("\nprinting only columns 0 and 2")
print(sales[[0,2]])
```
