Quick memory trick
Function	Question
np.any()	Is at least one True?
np.all()	Are all True?
axis=0	Work down columns
axis=1	Work across rows

```python
import numpy as np

sales = np.array([
    [120, 150, 180, 200],   # Product A
    [300, 280, 320, 350],   # Product B
    [500, 450, 470, 520],   # Product C
    [250, 270, 260, 290]    # Product D
])

print("Task 1")
result_1 = np.any(sales >500)
print(result_1)

print("\nTask 2")
result_2 = np.all(sales > 100)
print(result_2)

print("\nTask 3")
result_3 = np.any(sales > 400, axis = 1)
print(result_3)

print("\nTask 4")
result_4 = np.all(sales > 200, axis = 1)
print(result_4)

print("\nTask 5")
result_5 = np.any(sales > 400, axis = 0)
print(result_5)

print("\nTask 6")
result_6 = np.all(sales > 100, axis= 0)
print(result_6)
```
