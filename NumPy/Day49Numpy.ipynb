```python
# EXERCISE 2
import numpy as np

scores = np.array([
    [75, 80, 90],
    [60, 70, 85],
    [95, 88, 92]
])

print("Average of each student")
print(np.mean(scores, axis= 1))

print("\nAverage of each exam")
print(np.mean(scores, axis= 0))

print("\nHighest score of each student")
print(np.max(scores, axis= 1))

print("\nLowest score of each exam")
print(np.min(scores, axis= 0 ))
# Print:
# Average of each student
# Average of each exam
# Highest score of each student
# Lowest score of each exam
```

```python
# 📝 Exercise 3

import numpy as np

sales = np.array([
    [120, 450, 700],
    [900, 250, 180],
    [550, 610, 90],
    [800, 420, 300]
])

print("Total sales:")
print(np.sum(sales))

print("\nSales for each month")
print(np.sum(sales, axis= 0))

print("\nSales for each store")
print(np.sum(sales, axis= 1))

print("\nHighest sales")
print(np.max(sales))

print("\nLowest sales")
print(np.min(sales))

print("\nAverage sales")
print(np.mean(sales))

print("\n Standard deviation")
print(np.std(sales))
# Print:
# Total sales
# Sales for each month
# Sales for each store
# Highest sale
# Lowest sale
# Average sale
# Standard deviation
```

```python
# MINI CHALLENGE

import numpy as np

profits = np.array([
    [200, 350, 400],
    [500, 450, 600],
    [300, 700, 650],
    [800, 550, 900]
])

print("Total profit:")
print(np.sum(profits))

print("\nAverage profit per month")
print(np.mean(profits, axis= 0))

print("\nHighest profit for each company")
print(np.max(profits, axis= 1))

print("\nLowesh profit for each company")
print(np.min(profits, axis= 1))

print("\nCompany having highest total profit")
company_totals = np.sum(profits, axis = 1)
print(company_totals)

print("\nCompany having highest average profits")
monthly_average = np.mean(profits, axis=0)
print(monthly_average)

print("\nOverall Standard deviation")
print(np.std(profits))

# Find:
# 1. Total profit
# 2. Average profit per month
# 3. Highest profit for each company
# 4. Lowest profit for each company
# 5. Which company has the highest total profit?
# 6. Which month has the highest average profit?
# 7. Overall standard deviation
```
