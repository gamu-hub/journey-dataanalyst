# Lesson 1: Fancy Indexing

# Instead of slicing consecutive values, we can select any positions we want.

```python
import numpy as np

numbers = np.array([10,20,30,40, 50,60])
print(numbers[[0,2,5]])
```

```python
names = np.array([
    "John",
    "Mike",
    "Sarah",
    "David",
    "Emma"
])

print(names[[4,0,2]])
```

# Lesson 2: Fancy Indexing on 2D Arrays

```python
import numpy as np

scores = np.array([
    [80,90,75],
    [60,95,88],
    [100,70,65]
])

print(scores[[0,2]])

print("\nMultiple columns")
print(scores[:, [0,2]]) 
#: means every row.
#[0,2] emans Column 1 and Column 3
```

# Lesson 3: Selecting Individual Elements

Suppose we only want

80
95
65

```python
import numpy as np

scores = np.array([
    [80,90,75],
    [60,95,88],
    [100,70,65]
])

rows = [0,1,2]
cols = [0,1,2]

print(scores[rows, cols])
```

# Lesson 4: np.where()

One of the most useful NumPy functions.

```python
numbers = np.array([10, 25, 30, 18, 45])
print(np.where(numbers > 20))

# You can retrieve the values as well

indexes = np.where(numbers > 20)
print(numbers[indexes])
```

# Lesson 5: Replace Values Using np.where()

```python
numbers = np.array([10,25,30,18,45])

new_numbers = np.where(numbers > 20, 100, numbers)
print(new_numbers)

#Meaning

# If value > 20

# replace with 100

# otherwise keep original
```

# Lesson 6: Multiple Conditions

```python
numbers = np.array([10,25,30,18,45])
print(numbers[(numbers >20) & (numbers < 40)])
```
