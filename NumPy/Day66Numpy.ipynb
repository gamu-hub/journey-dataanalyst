# NumPy — np.any() and np.all() in practice

Yesterday, Day 65, you learned:

np.any() → checks whether at least one value is True
np.all() → checks whether every value is True
Using axis=0 → work down the rows / by column
Using axis=1 → work across the columns / by row

Today we'll build on that with practical filtering and combining conditions.

```python
import numpy as np
sales = np.array([
    [120, 150, 180, 200],
    [300, 280, 320, 350],
    [500, 450, 470, 520],
    [250, 270, 260, 290]
])

print("Task 1")
result_1 = np.any(sales > 400, axis = 0)
print(result_1)

print("\nTask 2")
result_2 = np.all(sales > 200, axis= 1)
print(result_2)

print("\nTask 3")
result_3 = np.any(sales < 130, axis = 1)
print(result_3)

print("\nTask 4")
result_4 = np.all(sales >= 250, axis = 1)
print(result_4)

print("\nTask 5")
result_5 = np.any(sales > 400, axis= 1) & np.any(sales < 300, axis= 1)
print(result_5)

print("\nTask 6")
result_6 = np.all(sales >= 200) & np.all(sales <= 500)
print(result_6)


print("\nTask 6 better corrected")

result_6 = np.all(
    (sales >= 200) & (sales <= 500),
    axis = 1
)
print(result_6)
```
