# 2. Selecting multiple columns

# 3. Selecting specific cells

You can select individual cells:

```python
import numpy as np

sales = np.array([
    [120, 150, 180, 200],
    [300, 280, 320, 350],
    [500, 450, 470, 520],
    [250, 270, 260, 290]
])

print(sales[2,2])
```

# But what if you want several specific cells?

For example:

row 0, column 1 → 150
row 1, column 3 → 350
row 3, column 0 → 250

```python
import numpy as np

sales = np.array([
    [120, 150, 180, 200],
    [300, 280, 320, 350],
    [500, 450, 470, 520],
    [250, 270, 260, 290]
])

rows = [0, 1, 3]
cols = [1, 3, 0]

print(sales[rows, cols])
```

# Your Day 58 Exercise

```python
import numpy as np

sales = np.array([
    [120, 150, 180, 200],
    [300, 280, 320, 350],
    [500, 450, 470, 520],
    [250, 270, 260, 290]
])

print("Row 1 and row 3")
print(sales[0:2])

print("\nPrinting column 1 and column 3")
print(sales[:,[0,2]])

print("\nPrinting specific values using adv indexing")
rows = [0,2,3]
cols = [2,1,3]
print(sales[rows, cols])

print("\nAll sales values greater than 400")
print(sales[sales > 400])

print("\nAll values that are greater than 250 AND less than 500")
print(sales[(sales > 250) & (sales < 500)])
```
