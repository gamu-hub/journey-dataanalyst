1. np.clip()

np.clip() forces values to stay within a specified range.

```python
import numpy as np

sales = np.array([100, 250, 400, 550, 700])

result = np.clip(sales, 200, 500)

print(result)
```

2. np.clip() with a 2D array

```python
import numpy as np
sales = np.array([
    [120, 150, 180, 200],
    [300, 280, 320, 350],
    [500, 450, 470, 520],
    [250, 270, 260, 290]
])

result = np.clip(sales, 200, 500)

print(result)

# Notice:

# Everything below 200 became 200
# Everything above 500 became 500
# Everything between 200 and 500 stayed unchanged
```

3. Why is this useful?

Imagine you have customer ages:

ages = np.array([5, 18, 25, 40, 67, 90, 120])

Maybe your analysis only accepts ages between 18 and 100.

clean_ages = np.clip(ages, 18, 100)

print(clean_ages)

Result:

[18 18 25 40 67 90 100]

This is a simple form of data cleaning.

4. Important distinction

Remember your NumPy symbols:

()

Used to call a function:

np.clip(sales, 200, 500)
[]

Used to create/access arrays and elements:

sales[0]
sales[:, 1]

So:

np.clip()

means run the clip function.
