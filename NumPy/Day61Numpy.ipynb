1. What is np.where()?

You already know how to filter:

sales[sales > 300]

This gives you the values that satisfy the condition.

But np.where() lets you say:

If this condition is true, give me this. Otherwise, give me that.

```python
import numpy as np

sales = np.array([100, 250, 400, 150, 500])

result = np.where(sales > 300, "High", "Low")

print(result)
```

2. np.where() can return positions

If you give it only a condition:

```python
sales = np.array([100, 250, 400, 150, 500])

positions = np.where(sales > 300)

print(positions)
```

3. np.where() with 2D arrays

Let's use our familiar sales dataset:

```python
sales = np.array([
    [120, 150, 180, 200],
    [300, 280, 320, 350],
    [500, 450, 470, 520],
    [250, 270, 260, 290]
])

# Suppose:

# Sales above 300 = "High"
# Everything else = "Low"
result = np.where(sales > 300, "High", "Low")
print(result)
```

4. np.where() vs Boolean Filtering

This distinction is important.

```python
sales[sales > 300]
```

```python
np.where(sales > 300, "High", "Low")
```

5. Multiple conditions

You can combine conditions.
For example:

Above 300 → "High"
Between 200 and 300 → "Medium"
Below 200 → "Low"

```python
# YOU CAN DO THIS WITH NESTED np.where()

result = np.where(
    sales > 300,
    "High",
    np.where(sales >= 200, "Medium", "Low")
)

print(result)
```
