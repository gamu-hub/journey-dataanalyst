1️⃣ .reshape()

reshape() changes the shape of an array.

```python
import numpy as np

sales = np.array([100, 200, 300, 400, 500, 600])

print(sales.reshape(2, 3))
```

2. Reshaping a 2D array

```python
sales = np.array([
    [100, 200, 300],
    [400, 500, 600]
])

print(sales.reshape(3, 2))
```

3. .flatten()
flatten() converts a multidimensional array into one dimension

```python
sales = np.array([
    [100, 200, 300],
    [400, 500, 600]
])

print(sales.flatten())
```

4. .ravel()
ravel() also converts an array into one dimension

```python
sales = np.array([
    [100, 200, 300],
    [400, 500, 600]
])

print(sales.ravel())
```

5️⃣ Transpose .T

This is very important for data analytics.

Transpose switches:

rows ↔ columns

```python
sales = np.array([
    [100, 200, 300],
    [400, 500, 600]
])

print(sales.T)
```
