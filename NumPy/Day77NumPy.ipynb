1️⃣ Boolean Indexing — The Core Idea

```python
import numpy as np

sales = np.array([120, 250, 180, 400, 90])

print(sales > 200)
```

# Filtering with Multiple Conditions

Suppose:

sales = np.array([120, 250, 180, 400, 90])

We want:

Sales greater than 100 AND less than 300.

Use &:

result = sales[(sales > 100) & (sales < 300)]

print(result)

```python
import numpy as np

result = sales[(sales > 100) & (sales < 300)]
print(result)
```

# Important numpy meaning.

- AND = &
- or = |
- NOT = ~
Suppose we want everything that is NOT greater than 200.

result = sales[~(sales > 200)]

print(result)


```python
result = sales[~(sales > 200)]
print(result)
```

Boolean Indexing with 2D Arrays

Now let's make this more realistic.

```python
sales = np.array([
    [120, 150, 180, 200],
    [300, 200, 320, 350],
    [90, 110, 130, 100]
])

print(sales > 200)

print(sales[sales > 200]) # to filter
```

np.where() — Finding Positions

This is extremely useful.

```python
sales = np.array([120, 250, 180, 400, 90])

positions = np.where(sales > 200)

print(positions)
```

np.where() Can Also Replace Values

This is very useful in data cleaning.

```python
sales = np.array([120, 250, 180, 400, 90])
```

We want:

Any value below 150 should become 150.

```python
result = np.where(sales < 150, 150, sales)
print(result)
```
