1️⃣ Basic Indexing

Let's start with a simple array:

```python
import numpy as np

sales = np.array([
    [120, 150, 180, 200],
    [300, 280, 320, 350],
    [500, 450, 470, 520],
    [250, 270, 260, 290]
])

print(sales[0, 0])
```

2️⃣ Selecting an Entire Row

```python
import numpy as np

sales = np.array([
    [120, 150, 180, 200],
    [300, 280, 320, 350],
    [500, 450, 470, 520],
    [250, 270, 260, 290]
])

print(sales[0])
print(sales[2])
```

3️⃣ Selecting a Column

```python
import numpy as np

sales = np.array([
    [120, 150, 180, 200],
    [300, 280, 320, 350],
    [500, 450, 470, 520],
    [250, 270, 260, 290]
])

print(sales[:, 0]) #:  → every row
```

4️⃣ Slicing

You can select multiple rows or columns.

First two rows

```python
import numpy as np

sales = np.array([
    [120, 150, 180, 200],
    [300, 280, 320, 350],
    [500, 450, 470, 520],
    [250, 270, 260, 290]
])

print(sales[0:2]) #mean Start at 0, stop before 2.
```

5️⃣ Boolean Conditions

Now things get interesting.

```python
import numpy as np

sales = np.array([
    [120, 150, 180, 200],
    [300, 280, 320, 350],
    [500, 450, 470, 520],
    [250, 270, 260, 290]
])

print(sales > 300)
```

6️⃣ Boolean Masking ⭐

```python
import numpy as np

sales = np.array([
    [120, 150, 180, 200],
    [300, 280, 320, 350],
    [500, 450, 470, 520],
    [250, 270, 260, 290]
])

print(sales[sales > 300])
```

7️⃣ Another Example

Find all sales greater than 400:

```python
import numpy as np

sales = np.array([
    [120, 150, 180, 200],
    [300, 280, 320, 350],
    [500, 450, 470, 520],
    [250, 270, 260, 290]
])

print(sales[sales > 400])
```

8️⃣ Filtering With <

Find all sales below 200:

```python
import numpy as np

sales = np.array([
    [120, 150, 180, 200],
    [300, 280, 320, 350],
    [500, 450, 470, 520],
    [250, 270, 260, 290]
])

print(sales[sales < 200])
```

9️⃣ Filtering With >=

Find sales of at least 300:

```python
import numpy as np

sales = np.array([
    [120, 150, 180, 200],
    [300, 280, 320, 350],
    [500, 450, 470, 520],
    [250, 270, 260, 290]
])

print(sales[sales >= 300])
```

🔟 Combining Conditions

You can combine conditions using:

```python
# Values between 200 and 400

print(sales[(sales >= 200) & (sales <= 400)])
```

# Day 55 Mini Challenge

```python
import numpy as np

sales = np.array([
    [120, 150, 180, 200],
    [300, 280, 320, 350],
    [500, 450, 470, 520],
    [250, 270, 260, 290]
])

print("printing only third row")
print(sales[2])

print("\nprinting the second column")
print(sales[:, 1])

print("\nAll values less than or equal to 250")
print(sales[sales <= 250])

print("\nAll values between 250 and 500 including 250 and 500")
print(sales[(sales >= 250) & (sales <= 500)])

print("\nValues greater than 300")
print(len(sales[sales > 300]))

print("\nReplacing every value greater than 400 with 999")
sales[sales > 400] = 999

print(sales)
```

Day 55 — Final Challenge

```python
import numpy as np

sales = np.array([
    [120, 150, 180, 200],
    [300, 280, 320, 350],
    [500, 450, 470, 520],
    [250, 270, 260, 290]
])

print("Greater than 250 AND less than 500")
print(sales[(sales > 250) & (sales < 500)])

print("\nHow many values satisfy that condition")
print(len(sales[(sales > 250) & (sales < 500)]))

print("\nReplace every value below 200 with 0")
sales[sales < 200] = 0
print(sales)

print("\nPrinting third row and replacing every value in that row that is greater than 480 with 1000")
sales[2][sales[2] > 480] = 1000
print(sales)

print("\nCombine challenge")
sales[sales >= 500] = 1000
sales[sales < 200] = 0
print(sales)

```
