```python
# 1️⃣ np.concatenate()

# Used to join arrays together.

# 1D example


import numpy as np

a = np.array([10, 20, 30])
b = np.array([40, 50, 60])

result = np.concatenate((a, b))

print(result)
```

2️⃣ Concatenating 2D arrays

```python
a = np.array([
    [10, 20],
    [30, 40]
])

b = np.array([
    [50, 60],
    [70, 80]
])
```

```python
# Join rows

result = np.concatenate((a, b), axis=0)

print(result)
```

# Join columns
result = np.concatenate((a, b), axis=1)

print(result)

```python
# Join columns
a = np.array([
    [10, 20],
    [30, 40]
])

b = np.array([
    [50, 60],
    [70, 80]
])

result = np.concatenate((a, b), axis=1)

print(result)
```

3️⃣ np.vstack()

```python
a = np.array([10, 20, 30])
b = np.array([40, 50, 60])

result = np.vstack((a, b))

print(result)
```

4️⃣ np.hstack()

hstack means horizontal stack.

```python
a = np.array([10, 20, 30])
b = np.array([40, 50, 60])

result = np.hstack((a, b))

print(result)
```

5️⃣ np.split()

This does the opposite of combining.

It breaks an array into smaller arrays.

```python
numbers = np.array([10, 20, 30, 40, 50, 60])

result = np.split(numbers, 3)

print(result)

#We split 6 numbers into 3 equal parts.
```
