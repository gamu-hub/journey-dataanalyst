Exercise 1: Converting Lists into Dictionaries
Key Python Topics:

Creating dictionaries
Zip function or dictionary comprehension


Instructions
You are given two lists. Convert them into a dictionary where the first list contains the keys and the second list contains the corresponding values.



```python
keys = ["Ten", "Twenty", "Thirty"]
values = [10, 20, 30]

#converting the lists into a dictionary
result = dict(zip(keys, values)) # the zip () method. 

print(result)
```

Exercise 2: Cinemax #2
Key Python Topics:

Looping through dictionaries
Conditionals
Calculations


Instructions
Write a program that calculates the total cost of movie tickets for a family based on their ages.

Family members’ ages are stored in a dictionary.
The ticket pricing rules are as follows:
Under 3 years old: Free
3 to 12 years old: $10
Over 12 years old: $15


Family Data:

family = {"rick": 43, 'beth': 13, 'morty': 5, 'summer': 8}


Loop through the family dictionary to calculate the total cost.
Print the ticket price for each family member.
Print the total cost at the end.


```python
family = {"rick": 43, "beth": 13, "morty": 5, "summer": 8}
total_cost = 0

for name, age in family.items(): # items() access both keys and values
    if age < 3:
        ticket_price = 0
    elif age < 13:
        ticket_price = 10

    else:
        ticket_price = 15

    print(name, ticket_price)

    total_cost += ticket_price # adding total cost and current price
print("Total cost is", total_cost)


```

Exercise 3: Zara
Key Python Topics:

Creating dictionaries

Dictionary methods like .pop() and .update()


Instructions
Create and manipulate a dictionary that contains information about the Zara brand.

Brand Information:

name: Zara
creation_date: 1975
creator_name: Amancio Ortega Gaona
type_of_clothes: men, women, children, home
international_competitors: Gap, H&M, Benetton
number_stores: 7000
major_color: 
    France: blue, 
    Spain: red, 
    US: pink, green
    
Create a dictionary called brand with the provided data.
Modify and access the dictionary as follows:
Change the value of number_stores to 2.
Print a sentence describing Zara’s clients using the type_of_clothes key.
Add a new key country_creation with the value Spain.
Check if international_competitors exists and, if so, add “Desigual” to the list.
Delete the creation_date key.
Print the last item in international_competitors.
Print the major colors in the US.
Print the number of keys in the dictionary.
Print all keys of the dictionary.

Bonus:

Create another dictionary called more_on_zara with creation_date and number_stores. Merge this dictionary with the original brand dictionary and print the result.


```python
# {creating an empty dictionary}

brand = {
    "name" : "Zara",
    "creation_date" : 1975,
    "creator_name" : "Aman Ortega Gaona",
    "type_of_clothes" : "men, women, children, home",
    "international_competitors" : ["Gap", "H&M", "Benetton"], # it contain a list
    "number_stores" : 7000,
    "major_color" : {
        "France" : "blue",
        "Spain" : "Red",
        "US" : [ "pink", "green"]  # contain a list
    }
}

print(brand)

print("changing number of stores 7000 to 2")
brand["number_stores"] = 2
print(brand["number_stores"])

print("\nZara's clients using type of clothes key")
print(f"Zara sells clothes for {brand["type_of_clothes"]}")

print("\nAdding new key country_cration with the value Spain")
brand["country_creation"] = "Spain"
print(brand["country_creation"])

print("\ncheck if international_competitors exists")
if "international_competitors" in brand:
    brand["international_competitors"].append("Disigual")
print(brand["international_competitors"])

print("\nDelete the creation_date key.")
brand.pop("creation_date")
print(brand)

print("\nPrinting the last item in international_competitors.")
print(brand["international_competitors"][-1])

print("Printing major colors in the US", brand["major_color"]["US"])

print("\nPrinting the number of keys in the dictionary")
print(len(brand))

print("\nPrint all keys of the dictionary.")
print(brand.keys())


# Second dictionary
more_on_zara = {
    "creation_date" : 1975,
    "number_of_stores" : 7000

}

# merging the new creations with brand
brand.update(more_on_zara)


print("Printing the final brand after merging", brand)

```

```python

users = ["Mickey", "Minnie", "Donald", "Ariel", "Pluto"]

# Dictionary 1: Characters mapped to indices
characters_to_indices = {}

for index, character in enumerate(users):
    characters_to_indices[character] = index

print("Dictionary 1:")
print(characters_to_indices)


# Dictionary 2: Indices mapped to characters
indices_to_characters = {}

for index, character in enumerate(users):
    indices_to_characters[index] = character

print("\nDictionary 2:")
print(indices_to_characters)


# Dictionary 3: Sorted characters mapped to indices
sorted_users = sorted(users)

sorted_characters_to_indices = {}

for index, character in enumerate(sorted_users):
    sorted_characters_to_indices[character] = index

print("\nDictionary 3:")
print(sorted_characters_to_indices)
```
