# Exercise 2: Birthdays Advanced
Instructions
Before asking the user to input a person’s name print out all of the names in the dictionary.
If the person that the user types is not found in the dictionary, print an error message (“Sorry, we don’t have the birthday information for <person’s name>”)


```python
birthdays = {
    "Lotzempa": "1992/08/04",
    "Niangboi": "1995/02/15",
    "Moimoi": "2024/11/27",
    "Naosen": "2026/10/28",
    "Sangsang": "2018/03/04"
}

# printing welcome messages
print("Welcome to the birthday look up")
print("You can look up the birthdays of the people from the list")

# asking user to give a person's name 
people = input("Enter a person's name: ")

# Checking whether the name exists in the dictionary
if people in birthdays:
    # Get the person's birthday
    birthday = birthdays[people]

    # printing a nice messages with f-string
    print(f"{people}'s birthday is on {birthday}")
else:
    # Print a message if the name isn't found
    print(f"Sorry, we don't have the birthday information for {people}")
```
