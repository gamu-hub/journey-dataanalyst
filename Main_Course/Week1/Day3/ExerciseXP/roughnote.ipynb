```python
my_string = '12345'
my_list = []

for c in my_string:
    my_list.append(int(c))
print(my_list)

# we can also do like this and its shorter
my_list2 = [int(c) for c in my_string]
print(my_list2)

my_new_list = []
for i in range(5):
    my_new_list.append(i * 2)
print(my_new_list)

my_new_list2 = [i * 2 for i in range(5)]
print(my_new_list2)

my_new_list3 = [i * 2 for i in range(5) if i % 2 == 0]
print(my_new_list3)
```
