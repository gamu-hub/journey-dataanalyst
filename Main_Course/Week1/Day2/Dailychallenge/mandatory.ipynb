1. Ask the user for two inputs:

A number (integer).
A length (integer).
2. Create a program that generates a list of multiples of the given number.
3. The list should stop when it reaches the length specified by the user.

Example 1:

Input:

number: 7
length: 5
Output:
[7, 14, 21, 28, 35]



```python
try:
    # Ask the user for a number and convert it to an integer.
    number = int(input("Enter a number: "))

    # Ask how many multiples the user wants.
    length = int(input("Enter the length: "))

    # Create an empty list to store the multiples.
    numbers = []

    # Repeat the loop as many times as the requested length.
    for i in range(length):
        # Calculate the next multiple.
        multiple = number * (i + 1)

        # Add the multiple to the list.
        numbers.append(multiple)

    # Display the completed list.
    print(numbers)

except ValueError:
    print("Please enter valid integers.")
```

1. Ask the user for a string.
2. Write a program that processes the string to remove consecutive duplicate letters.

The new string should only contain unique consecutive letters.
For example, “ppoeemm” should become “poem” (removes consecutive duplicates like ‘pp’, ‘ee’, and ‘mm’).
3. The program should print the modified string.

```python
# Step 1: Ask the user to enter a string
user_string = input("Enter a string: ")

# Step 2: Create an empty string to store the result
new_string = ""

# Step 3: Go through the original string one character at a time
for letter in user_string:

    # Step 4: If the new string is empty, add the first character
    if new_string == "":
        new_string = new_string + letter

    # Step 5: If the current letter differs from the last saved letter,
    # add it to the new string
    elif letter != new_string[-1]:
        new_string = new_string + letter

# Step 6: Print the modified string
print(new_string)
```
