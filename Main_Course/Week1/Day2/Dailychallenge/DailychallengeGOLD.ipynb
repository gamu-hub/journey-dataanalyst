Instructions
Ask the user for their birthdate (specify the format, for example: DD/MM/YYYY).
Display a little cake as seen below:
       ___iiiii___
      |:H:a:p:p:y:|
    __|___________|__
   |^^^^^^^^^^^^^^^^^|
   |:B:i:r:t:h:d:a:y:|
   |                 |
   ~~~~~~~~~~~~~~~~~~~

The number of candles on the cake should be the last number of the users age, if they are 53, then add 3 candles.

Bonus : If they were born on a leap year, display two cakes !

```python

# importing the tool for working with dates
from datetime import datetime, date

# Ask the user to enter their birth date
birthdate = input("Enter your birthday (DD/MM/YYYY): ")

# Converting the text into a date
birthdate = datetime.strptime(birthdate, "%d/%m/%Y")

# adding date on the import to get todays date
today= date.today()

# calculate the difference between the years
age = today.year - birthdate.year

# Check whether the birthday has happened this year. 
if (today.month, today.day) < (birthdate.month, birthdate.day):
    age = age - 1

# Display the age
print(f"You are {age} years old.")

# Get the last digit of the age
candles = age % 10

# Create the candles
candle_line = "i" * candles

# Print the birthday cake
print("       ___" + candle_line + "___")
print("      |:H:a:p:p:y:|")
print("    __|___________|__")
print("   |^^^^^^^^^^^^^^^^^|")
print("   |:B:i:r:t:h:d:a:y:|")
print("   |                 |")
print("   ~~~~~~~~~~~~~~~~~~~")
```
