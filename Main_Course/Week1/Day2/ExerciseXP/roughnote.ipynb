# 1. To create a list (let's call it "a") with The first 5 even numbers, starting from 2 
# print a list  that is created using this parttern [a[i]], a[i]**2, a [i+1], a[i+1]** 2, ....]



```python
# creating a list
a = [2, 4, 6, 8, 10]

# create an empty list for the result
result = []

# go throug each number 
for i in range(len(a)):

    # add the number to the result
    result.append(a[i])

    # add square of the number to the result 
    result.append(a[i]** 2)

print(sorted(result, reverse= True))
```
