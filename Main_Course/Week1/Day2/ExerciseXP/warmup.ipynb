

Today, we’re starting with a warm-up exercise.
Your task: Write a program that asks the user to enter a string.

If the string has an odd length: Reverse it and convert all characters to uppercase.
If the string has an even length: Create a new string of the same length, repeating the first and last characters of the original string (in lowercase) as many times as needed.
Print the resulting string.

```python

# Ask the user to enter a string
text = input("Enter a string: ")

# Check if the length of the string is odd
if len(text) % 2 != 0:

    # Reverse the string and convert it to uppercase
    result = text[::-1].upper()

# If the length of the string is even
else:

    # Get the first and last characters and convert them to lowercase
    first_char = text[0].lower()
    last_char = text[-1].lower()

    # Repeat the first and last characters to create a new string
    # The number of characters in the result must equal the original length
    result = (first_char + last_char) * (len(text) // 2)

# Print the resulting string
print(result)
```
