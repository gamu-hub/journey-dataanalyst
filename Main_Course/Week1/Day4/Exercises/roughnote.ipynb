```python

# Predict all the values in expressions.

expressions = {
    'booleans': [
        bool(None),
        'a' > '?',
        str(5) in '5',
        float(0.99) == int(0.99),
        not bool('A')
    ],
    'numbers': [
        (' ' in 'Hello World') * 3.5,
        True + 5,
        'Magical'.count('a') - 6.5,
        sum(x for x in range(11)),
        sum(x for x in range(3, 13, 3) if x % 3 < 2)
    ]
}

# 1. Count the number of True values in the booleans list.
true_count = expressions['booleans'].count(True)

# 2. Create a new list containing only positive numbers.
positive_numbers = [
    number
    for number in expressions['numbers']
    if number > 0
]

# 3. Calculate the sum of all positive numbers.
total = sum(positive_numbers)

# 4. Count how many positive numbers there are.
count = len(positive_numbers)

# 5. Calculate the average by dividing the sum by the count.
average = total / count

# 6. Print the results.
print("Number of True values:", true_count)
print("Average of positive numbers:", average)
```

```python
def say_hello(username, language = "en"):
    if language == "en":
        print(f"Hello {username}")
    elif language == "fr":
        print(f"Bonjour {username}")

    else:
        print("I'am sorry. I don't speak your language")

say_hello("John", "en")
say_hello("Mary", "fr")
say_hello("Bob", "he")

```

```python
# Adding more parameter to the above code

def say_hello(username, language = "en", message = ""):
    if language == "en":
        print(f"Hello {username} {message}")
    elif language == "fr":
        print(f"Bonjour {username} {message}")

    else:
        print("I'am sorry. I don't speak your language")

say_hello("John", "en")
say_hello("Mary", "fr")
say_hello("Bob", "he")
say_hello(username = "Adam", message = "Nice to meet you!")

```

```python


def my_sum(*args):
    print(f"My arg sum is ")
    sum = 0
    for i in args:
        sum += i
    return sum

print(my_sum(1,2,3,4,5,6))
```

```python
# using keyword argument (kwargs)

def my_sum(**kwargs):
    print(f"My args look like {kwargs}")
    print(kwargs ['username'])

print(my_sum(username = "Mary", age = 18, language = "en"))
```

```python
def calculation(*numbers, operation = "+"):

    #check which operation the user selected
    if operation == "+":
        # add all the nums together
        return sum(numbers)

    elif operation == "-":
        # Start with the first number
        result = numbers [0]

        # substract each remaining number
        for number in numbers[1:]:
            result -= number
        return result
# Calling the function with addition 
print(calculation(1, 2, 3))

# calling the function with subtraction

print(calculation(1,2,3, operation= "-"))
```

```python

# Define a function that adds 5 to a number
def add_five(number):
    return number + 5


# Create a list of numbers
numbers = [10, 20, 30]


# Use map() to apply add_five() to every number in the list
result = map(add_five, numbers)


# Convert the map object into a list and print the results
print(list(result))
```

```python
# USING LAMBDA


# Create a list of numbers
numbers = [10, 20, 30]

# Use lambda to define the operation directly inside map()
result = map(lambda number: number + 5, numbers)

# Convert the map object to a list and print it
print(list(result))
```

```python

# Create our own reduce function
def reduce(function, args):
    result = args[0]

    for e in args[1:]:
        result = function(result, e)

    return result


def calculation(*args, operation="+"):
    # Check that at least 2 numbers were provided
    if len(args) < 2:
        return "We need at least 2 numbers"

    # Addition
    if operation == "+":
        return sum(args)

    # Check whether the operation is valid
    if operation != "-":
        return "The operation is incorrect!"

    # Use reduce to subtract the numbers
    return reduce(lambda a, b: a - b, args)


# Test addition
print(calculation(1, 2, 3))

# Test subtraction
print(calculation(1, 2, 3, operation="-"))

# Test with only one number
print(calculation(1))
```
