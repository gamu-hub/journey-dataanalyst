Exercise 2: What’s Your Favorite Book?
Goal: Create a function that displays a message about a favorite book.

Key Python Topics:

Functions with parameters
String concatenation or f-strings
Calling functions with arguments


Step 1: Define a Function with a Parameter

Define a function named favorite_book().
This function should accept one parameter called title.


Step 2: Print a Message with the Title

The function needs to output a message like “One of my favorite books is <title>”.



Step 3: Call the Function with an Argument

Call the favorite_book() function and provide a book title as an argument.
For example: favorite_book("Alice in Wonderland").

```python
# define function = favorite_book and title is the parameter that receive the book title
def favorite_book(title):

    # Here we use f string so that we can insert the value of title
    print(f"One of my favorite book is {title}")

# calling the function with an argument aka call the function with a book title

favorite_book("Alice in wonderland")

```

Exercise 3: Some Geography
Goal: Create a function that describes a city and its country.

Key Python Topics:

Functions with multiple parameters
Default parameter values
String formatting


Step 1: Define a Function with Parameters ok

Define a function named describe_city().
This function should accept two parameters: city and country.
Give the country parameter a default value, such as “Unknown”.


Step 2: Print a Message

Inside the function, set up the code to display a sentence like “ is in “.
Replace <city> and <country> with the parameter values.


Step 3: Call the Function

Call the describe_city() function with different city and country combinations.
Try calling it with and without providing the country argument to see the default value in action.
Example: describe_city("Reykjavik", "Iceland") and describe_city("Paris").


Expected Output:

Reykjavik is in Iceland.
Paris is in Unknown.

```python
# define a function that should accept two parameters
# country has a default value of unknown. 

def describe_city(city, country = "Unknown"):

    # Diplay a sentence like "is in". We need to use f string since we have two values
    print(f"{city} is in {country}")

# call the function with both arguments
describe_city("Reykjavik", "Iceland")
describe_city("Paris")

```

Exercise 4: Random
Goal: Create a function that generates random numbers and compares them.

Key Python Topics:

random module
random.randint() function
Conditional statements (if, else)


Step 1: Import the random Module

At the beginning of your script, use import random to access the random number generation functions.


Step 2: Define a Function with a Parameter

Create a function that accepts a number between 1 and 100 as a parameter.


Step 3: Generate a Random Number

Inside the function, use random.randint(1, 100) to generate a random integer between 1 and 100.


Step 4: Compare the Numbers

If they are the same, print a success message. Otherwise, print a fail message and display both numbers.


Step 5: Call the Function

Call the function with a number between 1 and 100.


Expected Output:

Success! (if the numbers match)
Fail! Your number: 50, Random number: 23 (if they don't match)

```python
# random module

import random

# defining a function with a parameter

def compare_number(number):

    # using random.randint(1, 100)
    random_number = random.randint(1, 100)

# Checking if number exist in random_number
    if number == random_number:
        print("Success! ")

    # if not print the number follow by the value of number then Random number followed by the value of random_number    
    else:
        print(f"Your number: {number}, Random number: {random_number}")

# calling a function
compare_number(50)
```
