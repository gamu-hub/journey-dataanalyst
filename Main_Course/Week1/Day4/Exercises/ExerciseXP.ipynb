Exercise 4: Random
Goal: Create a function that generates random numbers and compares them.

Key Python Topics:

random module
random.randint() function
Conditional statements (if, else)


Step 1: Import the random Module

At the beginning of your script, use import random to access the random number generation functions.


Step 2: Define a Function with a Parameter

Create a function that accepts a number between 1 and 100 as a parameter.


Step 3: Generate a Random Number

Inside the function, use random.randint(1, 100) to generate a random integer between 1 and 100.


Step 4: Compare the Numbers

If they are the same, print a success message. Otherwise, print a fail message and display both numbers.


Step 5: Call the Function

Call the function with a number between 1 and 100.


Expected Output:

Success! (if the numbers match)
Fail! Your number: 50, Random number: 23 (if they don't match)

```python
# random module

import random

# defining a function with a parameter

def compare_number(number):

    # using random.randint(1, 100)
    random_number = random.randint(1, 100)

# Checking if number exist in random_number
    if number == random_number:
        print("Success! ")

    # if not print the number follow by the value of number then Random number followed by the value of random_number    
    else:
        print(f"Your number: {number}, Random number: {random_number}")

# calling a function
compare_number(50)
```

Exercise 6: Magicians…
Goal: Modify a list of magician names and display them in different ways.

Key Python Topics:

Lists
for loops
Modifying lists
Functions that modify data structures


Step 1: Create a List of Magician Names

Create a list called magician_names with the given names:
['Harry Houdini', 'David Blaine', 'Criss Angel']


Step 2: Create a Function to Display Magicians

Create a function called show_magicians() that takes the magician_names list as a parameter.
Inside the function, iterate through the list and print each magician’s name.


Step 3: Create a Function to Modify the List

Create a function called make_great() that takes the magician_names list as a parameter.
Inside the function, use a for loop to iterate through the list and add “the Great” before each magician’s name.


Step 4: Call the Functions

Call make_great() to modify the list.
Call show_magicians() to display the modified list.


Expected Output:

Harry Houdini the Great
David Blaine the Great
Criss Angel the Great

```python
# Step 1. Create a list named magician_names
magician_names = ["Harry Houdini", "David Blaine", "Criss Angel"]

# Step 2. Creating a function show_magicians that accepts the list as a parameter

def show_magicians(magician_names):

    # Instead of printing three different magicians names, for loop visits each item one at a time.
    for magician in magician_names:
        print(magician)

# Step 3. Creating a function to modify the list 

def make_great(magician_names):

    # Loop through the indexes of the list
    for i in range(len(magician_names)):
        # Add "the Great" to each name
        magician_names[i] = magician_names[i] + " the Great"


# Step 4. Modify the list before displaying it. 
make_great(magician_names)

# show the modified list
show_magicians(magician_names)



```
