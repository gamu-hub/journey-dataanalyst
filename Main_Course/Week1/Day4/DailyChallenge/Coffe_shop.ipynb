Daily Challenge: Coffee Shop Menu Manager
You were hired to help a small coffee shop manage their product menu using Python.

Write a program that:

1. Stores the coffee shop menu in memory
2. Lets the user:

Create a new item
Read (view) all items
Update an item’s price
Delete an item
Exit
Your program must be organized with functions.
Do not write all the logic in one giant while loop.
You should split behavior into reusable functions.

```python

# Coffee Shop Menu Manager

# Initial data: drink names and their prices
menu = {
    "espresso": 7.0,
    "latte": 12.0,
    "cappuccino": 10.0
}


# 1. Show all drinks and their prices
def show_menu(menu_dict):

    # Check if the dictionary is empty
    if not menu_dict:
        print("The menu is empty.")

    else:
        print("Current menu:")

        # Loop through every drink and its price
        for drink, price in menu_dict.items():
            print(f"{drink} - {price}₪")


# 2. Add a new drink
def add_item(menu_dict):

    # Ask the user for a drink name
    drink = input("Enter new drink name: ").lower()

    # Check whether the drink already exists
    if drink in menu_dict:
        print("Item already exists!")

    else:
        # Ask for the price and convert it to a float
        price = float(input("Enter price: "))

        # Add the drink and price to the dictionary
        menu_dict[drink] = price

        print(f'"{drink}" added!')


# 3. Update the price of an existing drink
def update_price(menu_dict):

    # Ask which drink the user wants to update
    drink = input("Which drink do you want to update? ").lower()

    # Check whether the drink exists
    if drink in menu_dict:

        # Ask for the new price
        new_price = float(input("Enter the new price: "))

        # Replace the old price with the new price
        menu_dict[drink] = new_price

        print("Price updated!")

    else:
        print("Item not found.")


# 4. Delete a drink
def delete_item(menu_dict):

    # Ask which drink the user wants to remove
    drink = input("Which drink do you want to delete? ").lower()

    # Check whether the drink exists
    if drink in menu_dict:

        # Remove the drink from the dictionary
        del menu_dict[drink]

        print("Item deleted.")

    else:
        print("Item not found.")


# 5. Display the available actions
def show_options():

    print("What would you like to do?")
    print("1. Show menu")
    print("2. Add item")
    print("3. Update price")
    print("4. Delete item")
    print("5. Exit")


# 6. Main controller: runs the coffee shop program
def run_coffee_shop():

    # Keep the program running until the user exits
    while True:

        # Display the available options
        show_options()

        # Ask the user to choose an option
        choice = input("Choose an option (1-5): ")

        # Call the correct function based on the choice
        if choice == "1":
            show_menu(menu)

        elif choice == "2":
            add_item(menu)

        elif choice == "3":
            update_price(menu)

        elif choice == "4":
            delete_item(menu)

        elif choice == "5":
            print("Goodbye!")
            break

        else:
            print("Invalid choice, try again.")


# Start the program
run_coffee_shop()
```
