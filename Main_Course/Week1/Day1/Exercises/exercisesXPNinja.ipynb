# Exercise 5: Longest word without a specific character
Instructions

1. Keep asking the user to input the longest sentence they can 

2. without the character “A”.
Each time a user successfully sets a new longest sentence, print a congratulations message.

```python
longest_sentence = 0

while True:
    sentence = input("Enter the longest sentence you can without the letter 'A': ")

    if "a" in sentence.lower():
        print("Sorry! The sentence contain the letter 'A'! Try again!!")

        continue

    current_length = len(sentence)

    
    if current_length > longest_sentence:
        longest_sentence = current_length
        print(f"Congratulations! {longest_sentence} characters")

    else:
        print(f"Good try! No 'A', but your sentence is {current_length} characters. Record is {longest_sentence}")

# I found one problem that the program keeps on printing so had to stop from IDE
```
