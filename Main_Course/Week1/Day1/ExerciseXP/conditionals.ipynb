A conditional lets python make a decision:
Suppose if something is true do this otherwise do someting else

```python
# The basic structure is:

# if condition:
    #code that runs if condition is True

age = 20

if age >= 18:
    print("You are an adult")
```

```python
# If.
# The condition must evaluate to either:
# True or False
```

```python

temperature = 30

if temperature > 25:
    print("It's hot")
```

```python
# else

age = 15
if age >= 18:
    print("Adult")
else:
    print("Minor")
```

```python
# elif

score = 75

if score >= 90:
    print("Execellent")
elif score >= 70:
    print("Good")
else:
    print("Needs improvement")
```

Comparison operators

These are extremely important:

Operator	Meaning	Example
==	equal to	x == 5
!=	not equal	x != 5
>	greater than	x > 5
<	less than	x < 5
>=	greater/equal	x >= 5
<=	less/equal	x <= 5

```python
# Exercise

# Create a variable: age = 25.
# Write a conditional that prints: You can vote.
# otherwise: You cannot vote

age = 25

if age >= 25:
    print("You can vote!")
else:
    print("You cannot vote")
```
