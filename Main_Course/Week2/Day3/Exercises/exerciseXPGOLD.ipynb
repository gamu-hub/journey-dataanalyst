```python

class Currency:
    def __init__(self, currency, amount):
        # Initialize the currency label and amount
        self.currency = currency
        self.amount = amount

    # Controls how the object is displayed with print()
    def __str__(self):
        return f"{self.amount} {self.currency}s"

    # Controls the official representation of the object
    def __repr__(self):
        return self.__str__()

    # Converts the Currency object into an integer
    def __int__(self):
        return self.amount

    # Handles the + operator
    def __add__(self, other):

        # Check if other is another Currency object
        if isinstance(other, Currency):

            # Check whether both currencies have the same label
            if self.currency == other.currency:
                return self.amount + other.amount

            # Different currency labels cannot be added
            else:
                raise TypeError(
                    f"Cannot add between Currency type <{self.currency}> "
                    f"and <{other.currency}>"
                )

        # If other is an integer, add it to the amount
        if isinstance(other, int):
            return self.amount + other

        # Reject unsupported types
        raise TypeError("Unsupported type for Currency addition")

    # Handles the += operator
    def __iadd__(self, other):

        # Check if other is another Currency object
        if isinstance(other, Currency):

            # Both currencies must have the same label
            if self.currency == other.currency:
                self.amount += other.amount
            else:
                raise TypeError(
                    f"Cannot add between Currency type <{self.currency}> "
                    f"and <{other.currency}>"
                )

        # If other is an integer, update the existing amount
        elif isinstance(other, int):
            self.amount += other

        # Reject unsupported types
        else:
            raise TypeError("Unsupported type for Currency addition")

        # Return the updated object
        return self


# Create Currency objects
c1 = Currency('dollar', 5)
c2 = Currency('dollar', 10)
c3 = Currency('shekel', 1)
c4 = Currency('shekel', 10)


# Test __str__()
print(c1)
# Expected: 5 dollars

# Test __int__()
print(int(c1))
# Expected: 5

# Test __repr__()
print(repr(c1))
# Expected: 5 dollars

# Test __add__() with an integer
print(c1 + 5)
# Expected: 10

# Test __add__() with another Currency
print(c1 + c2)
# Expected: 15

# Test that c1 remains unchanged after +
print(c1)
# Expected: 5 dollars

# Test __iadd__() with an integer
c1 += 5
print(c1)
# Expected: 10 dollars

# Test __iadd__() with another Currency
c1 += c2
print(c1)
# Expected: 20 dollars

# Test adding different currencies
# Uncomment this separately to test the TypeError
# print(c1 + c3)

# Test in-place addition with different currencies
c1 += c3
```
