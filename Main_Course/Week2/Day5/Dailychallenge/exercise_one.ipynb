Exercise 1: Quizz
Answer the following questions:

What is a class?
What is an instance?
What is encapsulation?
What is abstraction?
What is inheritance?
What is multiple inheritance?
What is polymorphism?
What is method resolution order or MRO?

1. A class is a blueprint or templates used to create objects.
2. An instance is an individual object created from a class.
3. It is the concept of grouping data or attributes and methods that operate on that data inside a class
4. Abstraction means hiding unnecessaty implementation details and exposing only the essential functionality. 
5. Inheritance allows a class to inherit attributes and methods from another class
6. Multiple inheritance occurs when a class inherits from more than one parent class
7. Polymorphism means many forms. It allows different classes to use the same method name but provide different implementation. 
8. MRO aka method resolution order determine the order in which Python searches parent classes for a method or attribute, particularly when multiple inheritance is used.
