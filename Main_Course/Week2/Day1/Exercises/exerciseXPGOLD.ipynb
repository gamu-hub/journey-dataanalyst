Exercise 1 : Geometry
Instructions
Write a class called Circle that receives a radius as an argument (default is 1.0).
Write two instance methods to compute perimeter and area.
Write a method that prints the geometrical definition of a circle.

```python
# Import Python's math module ( This is new to me but worth learning)
import math 

# Creating the Circle class

class Circle:

    # Initialize the circle with a default radius of 1.0
    def __init__(self, radius = 1.0):
        self.radius = radius

    #Calculate the perimeter
    def calculate_perimeter(self):
        return 2 * math.pi * self.radius # math.pi gives us the value pi

    # Calculate the area of the circle
    def calculate_area(self):
        return math.pi * self.radius ** 2


    # Print the geometrical difinition of a circle
    def print_definition(self):
        print("A circle is a set of all points in a plane"
              "that are at an equal distance from a fixed point"
              "called the center")
    
# TESTING THE CLASS
my_circle = Circle(5)

print("Perimeter:", my_circle.calculate_perimeter())
print("Area:", my_circle.calculate_area())
my_circle.print_definition()
```
