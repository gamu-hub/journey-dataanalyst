Step 1 — What is an instance method?
An instance method is a function defined inside a class that an object can use.

For example, a player has information such as their name and kills. They can also perform actions, such as introducing themselves.
- Attributes = information about an object.
- Methods = actions an object can perform.

```python
class Dog:
    def __init__(self, name_of_the_dog):
        self.name = name_of_the_dog

    def bark(self):
        print(f"{self.name} barks! WAF")

    def walk(self, number_of_metres):
        print(f"{self.name} walked {number_of_metres} metres")

    def rename(self, new_name):
        self.name = new_name


shelter_dog = Dog("Rex")
shelter_dog.rename("Paul")
print(shelter_dog.name)
shelter_dog.bark()
shelter_dog.walk(10)
```

```python
class Person():
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def show_details(self):
        print("Hello my name is " + self.name)

    def change_name(self, new_name):
        self.name = new_name

first_person = Person("John", 36)
first_person.show_details()

first_person.change_name ("David")
first_person.show_details()
```

Imagine we are building a simple banking system. Each bank account needs to store information and perform actions.

```python
class BankAccount:
    def  __init__(self, account_number, balance = 0): # this means if the caller provide a starting balance or if not python uses 0
        self.account_number = account_number
        self.balance = balance
        self.transactions = []

    def view_balance(self):
        self.transactions.append("View Balance")
        print(f"Balance for account {self.account_number}: {self.balance}")

    def deposit(self, amount):
        if amount <= 0:
            print("Inavalid amount")
        elif amount < 100:
            print("Minimum deposit is 100")
        else:
            self.balance += amount
            self.transactions.append(f"Deposit: {amount}")
            print("Deposit Successful")
            
    def withdraw(self, amount):
        if amount > self.balance:
            print("Insufficient Funds")
        else:
            self.balance -= amount
            self.transactions.append(f"Withdraw: {amount}")
            print("Withdraw Approved")
            return amount

    def view_transactions(self):
        print("Transactions:")
        print("-------------")
        for transaction in self.transactions:
            print(transaction)
```
