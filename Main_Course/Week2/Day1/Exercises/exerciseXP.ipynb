```python
class Cat:
    def __init__(self, cat_name, cat_age):
        self.name = cat_name
        self.age = cat_age

# Step 1 Create three cat objects with different names and ages
cat1 = Cat("Milo", 3)
cat2 = Cat("Luna", 5)
cat3 = Cat("Oliver", 2)


# Define a function that receive three cats objects
def find_oldest_cat(cat1, cat2, cat3):

    # Check if cat1 is older than both the second and third cats
    if cat1.age >cat2.age and cat1.age > cat3.age:

        # Return the first cat object if it is the oldest
        return cat1
    # If the first cat is not the oldest, check whether the second cat is older than the first and third cats
    elif cat2.age > cat1.age and cat2.age > cat3.age:

        # Return the second cat object if it is the oldest. 
        return cat2

    # If neither the first nor the second cat is the oldest, return the third cat
    else:
        return cat3

# Step 3: Call the function with the three cat object
# Store the oldest cat object returned by the function
oldest_cat = find_oldest_cat(cat1, cat2, cat3)

# Print the oldest cat's name and age using f-string
print(f"The oldest is {oldest_cat.name}, and is {oldest_cat.age} years old")
```
