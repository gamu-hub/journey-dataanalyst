```python
# Step 1: Create a zoo class

class Zoo:
    # initialize the zoo with its name
    def __init__(self, zoo_name):

        # Store the zoo's name
        self.zoo_name = zoo_name

        # Create an empty list to store animal names
        self.animals = []

    # Create a method to add an animal to the zoo
    def add_animal(self, new_animal):

        # Check if the animal is not already in the list
        if new_animal not in self.animals:

            # Add the new animals on the list
            self.animals.append(new_animal)

    # Create a method to display all animals in the zoo
    def get_animals(self):

        # Print the list of the animals
        print(self.animals)

    # Create a method to sell an animal
    def sell_animal(self, animal_sold):

        # Check if the animal exists in the zoo
        if animal_sold in self.animals:

            # Remove the animal from the list
            self.animals.remove(animal_sold)

    # Create a method to sort and group the animals

    def sort_animals(self):
        # Sort the animal names alphabetically
        self.animals.sort()

        # Create an empty dictionary for the animal groups
        groups = {}

        # Go through evevery animal in the list i.e self.animals
        for animal in self.animals:

            # Get the first letter of the animal's name
            first_letter = animal[0]

            # Check if this letter is not already in the dictionary
            if first_letter not in groups:

                # Create an empty list for this letter
                groups[first_letter] = []

            # Add the animal to its correct group
            groups[first_letter].append(animal)

        # Save the grouped animals in the Zoo object
        self.animal_groups = groups

    # Create a method to display animal groups
    def get_groups(self):

        # Go through each letter and its list of animals
        for letter, animals in self.animal_groups.items():

            #Print the letter and its animals
            print(f"{letter}: {animals}")

brooklyn_safari = Zoo("Brooklyn Safari")

brooklyn_safari.add_animal("Giraffe")
brooklyn_safari.add_animal("Bear")
brooklyn_safari.add_animal("Baboon")
brooklyn_safari.add_animal("Cat")

brooklyn_safari.get_animals()

brooklyn_safari.sell_animal("Bear")
brooklyn_safari.get_animals()

brooklyn_safari.sort_animals()
brooklyn_safari.get_groups()

brooklyn_safari.add_animal("Cat")
brooklyn_safari.get_animals()


        
```
