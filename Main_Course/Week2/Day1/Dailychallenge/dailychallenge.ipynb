Step 8: upgrade the add_animal Method

use **kwargs for passing multiple animals. The keys will be the animal name and the value will be the quantity.
Then you can call the method this way: macdonald.add_animal('cow'= 5, 'sheep' = 2, 'goat' = 12)

```python
# create a class called Farm
class Farm:

    # Initialize the farm
    def __init__(self, farm_name):

        # Store the farm name
        self.name = farm_name

        # Create an empty dictionary for animals
        self.animals = {}


    # Use a method to add more animals
    def add_animal(self, animal_type, count = 1): 

        # check animals exist in animal_type
        if animal_type in self.animals:

            # we increments its count by count since it exist
            self.animals [animal_type] += count

        # if the animal does not exist, we create a new animal
        else:
            self.animals[animal_type] = count

    # Return information about the farm
    def get_info(self):

        # Create the farm heading with a blank line
        info = f"{self.name}'s Farm\n\n"

        # Go through each animal and count
        for animal, count in self.animals.items():

            # Add each animal and count in alligned columns
            info += f"{animal:<10} : {count}\n"

        # Add the final farm message
        info += "\n E-I-E-I-0!"

        # Return the completed information
        return info


# TEST CODE — outside the class

macdonald = Farm("McDonald")

macdonald.add_animal("cow", 5)
macdonald.add_animal("sheep", 2)
macdonald.add_animal("goat", 12)

print(macdonald.get_info())
```
