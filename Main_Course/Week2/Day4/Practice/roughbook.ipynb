```python
""" Imagine you receive JSON data as a string
and want to work with it using Python"""

import json

team = {
    "name": "HOPE Esports",
    "kills": 85,
    "matches": 12,
    "qualified": True,
    "players": ["Alpha", "Bravo", "Charlie", "Delta"]
}

python_data = json.loads(json_data)
print(python_data)

print(type(python_data))

print(python_data["name"])
```

```python
with open("tournament.json", "w") as file:
    json.dump(team, file)
# What happened?
# 1. Opened tournament.json.
# 2. Used "w" to create or overwrite the file.
# 3. Used json.dump() to write our dictionary into the file.
# 4. The file is automatically closed.

with open("tournament.json", "r") as file:
    loaded_data = json.load(file)
print(loaded_data)
print(type(loaded_data))
print(loaded_data["name"])
print(loaded_data["kills"])
```
