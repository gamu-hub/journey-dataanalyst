Exercise 1 – Single Inheritance
Beginner level
Scenario: HOPE Esports Player Management

We have a parent class called Player and a child class called ProPlayer.

The parent class stores:
- Player name
- Team name
The child class adds:

- Tournament wins

```python
class Player:

    def __init__(self, name, team):
        self.name = name
        self.team = team

class ProPlayer(Player):

    def __init__(self, name, team, wins):
        super().__init__(name, team)
        self.wins = wins

player1 = ProPlayer("Alex", "HOPE", 5)

print("Player:", player1.name)
print("Team:",player1.team)
print("Tournament wins:", player1.wins)
        
```

Exercise on Inheritance and Composition: Door class
Try to recreate the class explained below:

We have a class called Door that has an attribute of is_opened declared when an instance is initiated.

We can create a class called BlockedDoor that inherits from the base class Door.

We just override the parent class's functions of open_door() and close_door() with our own BlockedDoor version of those functions, which simply raises an Error that a blocked door cannot be opened or closed.

```python

# Parent class
class Door:

    def __init__(self, is_opened):
        self.is_opened = is_opened

    def open_door(self):
        self.is_opened = True
        print("The door is now open.")

    def close_door(self):
        self.is_opened = False
        print("The door is now closed.")


# Child class
class BlockedDoor(Door):

    def open_door(self):
        raise Exception("A blocked door cannot be opened!")

    def close_door(self):
        raise Exception("A blocked door cannot be closed!")


# Create objects
door1 = Door(False)
blocked_door1 = BlockedDoor(False)


# Test normal door
door1.open_door()
print("Door status:", door1.is_opened)

door1.close_door()
print("Door status:", door1.is_opened)


# Test blocked door
try:
    blocked_door1.open_door()
except Exception as e:
    print("Error:", e)

try:
    blocked_door1.close_door()
except Exception as e:
    print("Error:", e)
```

```python
class Book:
    def __init__(self, title, author, publication_date, price):
        self.title = title
        self.author = author
        self.publication = publication_date
        self.price = price

    def present(self):
        print(f'Title: {self.title}')


class Fiction(Book):
    def __init__(self, title, author, publication_date, price, is_awesome):
        super().__init__(title, author, publication_date, price)

        self.genre = 'Fiction'
        self.is_awesome = is_awesome

        if self.is_awesome:
            self.bored = False
            print('Woow this is an awesome book')
        else:
            self.bored = True
            print('I am very bored')

foundation = Fiction('Asimov', 'Foundation', '1966', '5£', True)

print(foundation.title)
```
