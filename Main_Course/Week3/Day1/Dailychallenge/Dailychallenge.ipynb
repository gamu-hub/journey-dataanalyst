### Case Study Summary

FC Schalke 04 stored sports data and business data in separate systems. This made it difficult to combine and analyse the information.

The club used Microsoft Fabric and Azure Databricks to create a shared data platform.

The data integration helped the club:

- Make initial data integration 3 times faster.
- Reduce report creation time by 60%.
- Combine and clean 220,000 membership records.
- Complete the first use case in 2 months.

### Impact

Data integration reduced manual work and allowed the small data team to focus more on analysis.

Better business data can help the club understand its fans and identify revenue opportunities. Sports data can support scouting and player evaluation.
