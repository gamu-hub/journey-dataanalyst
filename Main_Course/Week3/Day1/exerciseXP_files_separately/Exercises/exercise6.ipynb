## Exercise 6: Identifying Data Types

1. **A company's financial reports stored in an Excel file:** Structured Data — Information is organised into rows and columns.

2. **Photographs uploaded to a social media platform:** Unstructured Data — Images do not follow a predefined table structure.

3. **A collection of news articles on a website:** Unstructured Data — Articles contain free-form text without a fixed table structure.

4. **Inventory data in a relational database:** Structured Data — Information is organised into tables with predefined columns and rows.

5. **Recorded interviews from a market research study:** Unstructured Data — Audio recordings do not follow a predefined table structure.
