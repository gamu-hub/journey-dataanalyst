Exercise 7: Transformation Exercise
For each of the following unstructured data sources, propose a method to convert it into structured data. Explain your reasoning.

A series of blog posts about travel experiences.
Audio recordings of customer service calls.
Handwritten notes from a brainstorming session.
A video tutorial on cooking.

## Exercise 7: Transformation Exercise

### 1. Travel Blog Posts

**Method:** Extract information such as country, destination, travel dates, and activities from the blog posts and organise it into an Excel spreadsheet.

**Reason:** This makes it easier to compare destinations and identify travel trends.

### 2. Audio Recordings of Customer Service Calls

**Method:** Use speech-to-text technology to convert audio recordings into written text. Then extract information such as customer problems, complaints, and satisfaction levels into a table.

**Reason:** This allows the company to identify common customer problems and analyse customer feedback.

### 3. Handwritten Notes from a Brainstorming Session

**Method:** Scan the handwritten notes and convert them into digital text using OCR (Optical Character Recognition). Then organise the ideas into categories in a spreadsheet.

**Reason:** This makes the ideas easier to search, organise, and compare.

### 4. Video Tutorial on Cooking

**Method:** Extract information such as ingredients, quantities, cooking steps, and cooking times from the video and organise them into a structured recipe table.

**Reason:** This makes the information easier to follow, search, and compare with other recipes.
