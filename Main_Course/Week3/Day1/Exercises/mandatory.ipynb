# Exercise 1: Introduction to Data Analysis

## 1. What Is Data Analysis?

Data analysis is the process of collecting, organising, cleaning and examining data to discover useful information, identify patterns and make better decisions.

In esports, data analysis can be used to understand team performance, player statistics and tournament results. For example, as the main organiser of a PUBG tournament, I can analyse match results, kills, placement points and total scores to determine team rankings and identify the strongest teams.

Instead of relying only on opinions, tournament organisers can use data to understand what actually happened during a competition.

## 2. Why Is Data Analysis Important in Modern Contexts?

Data analysis is important in the modern world because organisations collect large amounts of data every day. Analysing this data helps people identify patterns, solve problems, improve performance and make informed decisions.

In esports, data analysis helps tournament organisers manage competitions more effectively. By analysing registration numbers, match results, player statistics and audience engagement, organisers can understand what works well and what needs improvement.

For example, an organiser can compare viewer numbers across different tournament broadcasts to identify which matches attract the most attention. This information can help improve future events, attract sponsors and create better experiences for players and viewers.

Data analysis can also help esports teams evaluate their performance, identify weaknesses and develop better strategies for future matches.

## 3. Three Areas Where Data Analysis Is Applied Today

### A. Esports and Gaming

Data analysis is widely used in esports to evaluate team performance, analyse player statistics and improve competitive strategies.

For example, in a PUBG tournament, organisers can analyse kills, placement points, survival time and match results to calculate standings and compare team performance. Teams can also use these statistics to identify their strengths and weaknesses and improve their strategies.

### B. Marketing and Audience Engagement

Businesses use data analysis to understand their audiences, measure engagement and improve marketing campaigns. They analyse information such as views, clicks, registrations and social media interactions.

For example, an esports tournament organiser can analyse Instagram engagement, YouTube views and Discord activity to understand which announcements attract the most attention. These insights can help organisers promote future tournaments more effectively and demonstrate audience reach to potential sponsors.

### C. Finance and Tournament Management

Financial analysis helps organisations monitor expenses, manage budgets and evaluate profitability. It allows decision-makers to understand where money is being spent and whether an event is financially sustainable.

For example, when organising a PUBG tournament, I can analyse the prize pool, sponsorship income, registration fees and operating expenses. This information helps determine whether the tournament stayed within budget and how much funding is needed for future events.

## Conclusion

In conclusion, data analysis is an important tool for understanding information and making better decisions. It is useful in many industries, including esports, marketing and finance. Through my experience organising PUBG tournaments, I can see how analysing match statistics, audience engagement and financial records can improve tournament management, support better planning and create more successful esports events.
