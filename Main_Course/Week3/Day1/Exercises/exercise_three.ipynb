# Exercise 3: Identifying Data Types

## 1. Sleep Dataset

- index — Qualitative: Identifies each row.
- Year — Quantitative: Represents a numerical year.
- Period — Qualitative: Represents a time-period category.
- Avg hrs per day sleeping — Quantitative: Measures sleeping hours.
- Standard Error — Quantitative: Represents a numerical statistical measurement.
- Type of Days — Qualitative: Describes the type of day.
- Age Group — Qualitative: Represents age categories.
- Activity — Qualitative: Describes an activity.
- Sex — Qualitative: Represents categories.

## 2. Mental Health Dataset

- index — Qualitative: Identifies each row.
- Entity — Qualitative: Represents a country or entity name.
- Code — Qualitative: Identifies an entity.
- Year — Quantitative: Represents a numerical year.
- Schizophrenia (%) — Quantitative: Represents a percentage.
- Bipolar disorder (%) — Quantitative: Represents a percentage.
- Eating disorders (%) — Quantitative: Represents a percentage.
- Anxiety disorders (%) — Quantitative: Represents a percentage.
- Drug use disorders (%) — Quantitative: Represents a percentage.
- Depression (%) — Quantitative: Represents a percentage.
- Alcohol use disorders (%) — Quantitative: Represents a percentage.

## 3. Credit Card Dataset

- Gender — Qualitative: Represents a category.
- Age — Quantitative: Measures a person's age.
- Debt — Quantitative: Represents a numerical amount.
- Married — Qualitative: Represents a category.
- BankCustomer — Qualitative: Represents a customer category.
- Industry — Qualitative: Represents an industry category.
- Ethnicity — Qualitative: Represents a category.
- YearsEmployed — Quantitative: Measures years of employment.
- PriorDefault — Qualitative: Represents a yes/no category.
- Employed — Qualitative: Represents an employment category.
- CreditScore — Quantitative: Represents a numerical score.
- DriversLicense — Qualitative: Represents a yes/no category.
- Citizen — Qualitative: Represents a citizenship category.
- ZipCode — Qualitative: Identifies a location.
- Income — Quantitative: Represents a numerical amount.
- Approved — Qualitative: Represents an approval category.
