# Exercise 1: Introduction to Data Analysis

## 1. What Is Data Analysis?

Data analysis is the process of collecting, organising, examining, and interpreting data to find useful information. It helps people understand what happened, identify patterns, and make better decisions based on facts rather than guesses.

For example, as a PUBG tournament organiser, I can analyse match results, player rankings, kills, team performance, and tournament registrations to understand how well an event performed.

## 2. Why Is Data Analysis Important in Modern Contexts?

Data analysis is important because organisations generate large amounts of data every day. By analysing this information, they can identify problems, improve performance, understand people's behaviour, and make better decisions.

In esports, data analysis helps tournament organisers evaluate player performance, improve tournament formats, understand audience engagement, and attract sponsors. For example, by analysing YouTube views, Discord activity, and tournament participation, an organiser can understand which events attract the most interest.

Today, data analysis is also increasingly used with artificial intelligence and machine learning to identify patterns and predict possible outcomes. However, the quality of the data is important because incorrect or incomplete data can lead to poor decisions.

## 3. Three Areas Where Data Analysis Is Applied Today

### A. Esports and Gaming

Data analysis is used to evaluate players, teams, tournaments, and audience engagement. Organisers can examine kills, match placements, win rates, registrations, and viewership statistics.

For example, a PUBG tournament organiser can analyse match results to create accurate rankings, identify top-performing teams, and understand which tournaments attract more viewers. This information can also help when presenting performance reports to sponsors.

### B. Healthcare

Hospitals and healthcare organisations analyse patient information, medical test results, and treatment outcomes. This helps healthcare professionals identify trends, plan resources, and support better decisions about patient care.

For example, hospitals can analyse patient admission data to estimate how many beds and medical staff may be needed during busy periods.

### C. Banking and Finance

Banks and financial institutions use data analysis to understand spending patterns, assess credit applications, and detect potentially fraudulent transactions.

For example, a bank can analyse a customer's financial information to help assess whether the customer meets the requirements for a credit card. It can also identify unusual transaction patterns that may indicate fraud.

## Conclusion

Data analysis is an important skill in the modern world because it turns raw data into useful information. It helps people make informed decisions, solve problems, and improve performance. From organising PUBG tournaments to managing hospitals and preventing financial fraud, data analysis plays an important role in many industries.
