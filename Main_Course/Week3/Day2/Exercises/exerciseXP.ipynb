### Exercise 1: Duplicate Detection and Removal

I loaded the Titanic dataset and checked for duplicate rows using Pandas' `duplicated()` function. The dataset contained **891 rows and 0 duplicate rows**.

I then used `drop_duplicates()` to remove any duplicate rows and checked the number of rows again. The dataset still contained 891 rows, meaning that no rows were removed.

Finally, I verified that the dataset contained zero duplicate rows after the cleaning process.

**Conclusion:** The Titanic dataset did not contain any completely duplicated rows, so no rows needed to be removed.

```python
import pandas as pd

titanic_data = pd.read_csv("/content/train.csv")
titanic_data.head()
```

```python
# Count rows before cleaning
rows_before = len(titanic_data)

# Detect duplicate rows
duplicate_count = titanic_data.duplicated().sum()

print("Rows before:", rows_before)
print("Duplicate rows:", duplicate_count)

#Remove duplicates rows
titanic_data = titanic_data.drop_duplicates()

#Verify the result
rows_after = len(titanic_data)

print("Rows after:", rows_after)
print("Rows removed:", rows_before - rows_after)
print("Duplicates remaining:", titanic_data.duplicated().sum())
```

```python
# Since we already read the csv file we don't need to import again

# Lets count the missing values in each column
missing_values = titanic_data.isnull().sum()

print(missing_values)

```

```python
# Keep only columns with missing values

missing_values = missing_values[missing_values > 0]

#Display the results
print(missing_values)
```

```python
# Record the number of rows and columns before cleaning
# Create a copy for Exercise 2
titanic_cleaned = titanic_data.copy()
rows_before = len(titanic_cleaned)
columns_before = len(titanic_cleaned.columns)

print("Rows before:", rows_before)
print("Columns before:", columns_before)
```

```python
# Create a copy of the original dataset
titanic_cleaned = titanic_data.copy()

# Calculate the median age using the copied dataset
median_age = titanic_cleaned['Age'].median()

print("Median age:", median_age)
```

```python
# Fill missing Age values with the median

titanic_cleaned['Age'] = titanic_cleaned['Age'].fillna(median_age)
print("Missing Age values:", titanic_cleaned['Age'].isnull().sum())
```

```python
# Count the missing Cabin values
print("Missing Cabin Values:",
      titanic_cleaned['Cabin'].isnull().sum())
```

```python
# Count cabin values that are not missing
print("Available Cabin values:",
      titanic_cleaned['Cabin'].notnull().sum())
```

We will not change the missing cabin values for now because we can come back later. Because sometimes keeping missing values is better than introducing misleading information

```python
# Find the most frequent Embarked value
most_common_port = titanic_cleaned['Embarked'].mode()[0]

print("Most common port:", most_common_port)
```

```python
most_common_port = titanic_cleaned['Embarked'].mode()[0]
titanic_cleaned['Embarked'] = titanic_cleaned['Embarked'].fillna(most_common_port)
print("Missing Embarked values:", titanic_cleaned['Embarked'].isnull().sum())
```

```python
print(titanic_cleaned.isnull().sum())
```

```python
# We don't want to changed the cabin column rather we will verify my dataset size because
# we want to confirm that handling missing values didn't accidentally remove any passengers or columns.
print("Rows and columns:", titanic_cleaned.shape)
```

```python
# Handling missing values to final check
print(titanic_cleaned.isnull().sum().sum())
```
