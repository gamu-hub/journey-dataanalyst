```python
import pandas as pd
df = pd.read_csv(
    "/content/Data Science Job Salary dataset/datascience_salaries.csv"
)

# Display the first 5 rows
df.head()

# Display the Dataset information
print(df.info())

# Display the number of rows and columns
print(df.shape)

# Display for missing values
print(df.isnull().sum())
```

```python
# Checking Unnamed: 0 column, we need to inspect. If it is just a row number, we will remove it.
print(df["Unnamed: 0"].head())
```

```python
# Now we confirm Unnamed:0 is just a saved row index so we will now remove.
df = df.drop("Unnamed: 0", axis= 1)
```

```python
# MinMaxScaler() creates the normalization tool
# fit_transform() calculates the minimum and maximum salary,the scales the values.
# df["salary_normalized"] creates a new column to store the result

from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()
df["salary_normalized"] = scaler.fit_transform(df[["salary"]])
```

```python
# print the result to check
print(df[["salary", "salary_normalized"]].head())

# Checking the minimum
print(df["salary_normalized"].min())

# Checking the maximum
print(df["salary_normalized"].max())
```

```python
# Creating a separate DataFrame not to modify the original df because we will need later on.
pca_df = df.drop(columns=["salary"])

# Remember we remove original salary column since creating salary_normalized which contain the same salary info.
```

```python
# convert the text columns into numbers and run again PCA
pca_df = df.drop(columns=["salary"])
```

```python
# Converting the categorical columns into numerical columns
pca_df = pd.get_dummies(
    pca_df,
    columns= [
        "job_title",
        "job_type",
        "experience_level",
        "location",
        "salary_currency"
    ],
    dtype= int
)
```

```python
# Checking the result
print("Head")
print(pca_df.head())

print("\nshape")
print(pca_df.shape)
```

```python
# Applying PCA

from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
pca_scaled = scaler.fit_transform(pca_df)
```

```python
# Lets reduce the many columns into 2 principle components
from sklearn.decomposition import PCA
pca = PCA(n_components=2)
```

```python
# Now lets tranform the data
pca_result = pca.fit_transform(pca_scaled)
```

```python
# Create a DataFrame
pca_result_df = pd.DataFrame(
    pca_result,
    columns=["PC1", "PC2"]
)

print(pca_result_df.head())
print(pca_result_df.shape)
```

```python
# Lets check how much information PCA retained
print(pca.explained_variance_ratio_)

# This shows the proportion of variance explained by each principal component.
```

```python
# To calculate the total explained variance
print(pca.explained_variance_ratio_.sum())
```

```python
# GROUP DATA: We will use the original df not the PCA DataFrame

salary_by_experience = df.groupby("experience_level")["salary"].agg(
    ["mean", "median"]
)

print(salary_by_experience)
```

## Conclusion

In this exercise, I performed data preprocessing on the Data Science Job Salary dataset.

First, I applied Min-Max normalization to the salary column, scaling the salary values between 0 and 1.

Second, I used Principal Component Analysis (PCA) to reduce the number of features in the dataset to two principal components.

Finally, I grouped the data by experience level and calculated the average and median salary for each group.

The results showed that executive-level jobs had the highest average salary, while entry-level jobs had the lowest average and median salaries. Senior-level jobs also showed higher median salaries than mid-level jobs.

This exercise helped me understand how normalization, dimensionality reduction, and data aggregation can be used to prepare and analyze data effectively.
