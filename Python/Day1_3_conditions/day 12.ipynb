Why Do We Need Error Handling?

Without error handling:

age = int(input("Enter age: "))
print(age)

If the user enters:

abc

Python crashes with:

ValueError

The program stops.

That's bad if you're making a real application.

1. try and except
try:
    age = int(input("Enter age: "))
    print(age)

except ValueError:
    print("Invalid input")

    What happens?

If user enters:

25

Output:

25

If user enters:

abc

Output:

Invalid input

No crash. ✅

2. ValueError

Occurs when Python receives the wrong value type.

Example:

number = int("hello")

Error:

ValueError

Because "hello" cannot become an integer.



3. ZeroDivisionError

Example:

result = 10 / 0

Error:

ZeroDivisionError

Handle it like this:

```python
try:
    number = int(input("Enter number: "))
    result = 100 / number
    print(result)

except ZeroDivisionError:
    print("Cannot divide by zero")
```

4. FileNotFoundError
Example:

open("movies.txt", "r")
If the file doesn't exist:

FileNotFoundError

5. Multiple Exceptions

One try can have multiple except blocks.

```python
try:
    number = int(input("Enter number: "))
    result = 100 / number

except ValueError:
    print("Please enter a number")

except ZeroDivisionError:
    print("Cannot divide by zero")
```

```python
6. while True

Runs forever until break.

while True:
    print("Hello")

Output:

Hello
Hello
Hello
...
```

```python
7. break

Stops a loop.

while True:
    number = int(input("Number: "))

    if number > 0:
        break

print("Done")

If user enters:

-1
-5
3

Output:

Done

Loop stopped because of break.
```

8. Combining while True + try

This is the most important pattern from Day 12.




```python
while True:
    try:
        number = int(input("Enter number: "))
        print("Thank you")
        break

    except ValueError:
        print("Invalid")
```

What happens?

Enter number: hello
Invalid

Enter number: abc
Invalid

Enter number: 10
Thank you

This pattern is used everywhere in real programs.

9. Reading Errors

When Python shows:

ValueError: invalid literal for int()

Look at:

The error type (ValueError)
The line number
The code on that line

Ask:

Did I type text instead of a number?
Did I divide by zero?
Is the file missing?
Is a variable misspelled?
