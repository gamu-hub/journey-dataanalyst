1. OPENING A FILE
with open("fruits.txt", "r") as file:

Breakdown:

Part	Meaning
open()	Open a file
"fruits.txt"	File name
"r"	Read mode
as file	Store opened file in variable file

2. FILE MODES

These are the three most important modes.

Read Mode
"r"

Used when:

I want to look at data.

Example:

with open("fruits.txt", "r") as file:

Can:

read

Cannot:

write
Write Mode
"w"

Used when:

I want to replace everything.

Example:

with open("fruits.txt", "w") as file:

Can:

write

Danger:

Deletes old content.

Example:

File:

Apple
Banana

Code:

with open("fruits.txt", "w") as file:
    file.write("Orange")

File becomes:

Orange

Everything else is gone.

Append Mode
"a"

Used when:

I want to add new data.

Example:

with open("fruits.txt", "a") as file:

Existing data stays.

New data gets added.

3. WRITING TO A FILE

Example:

fruit = input("Favourite fruit: ")

with open("fruits.txt", "a") as file:
    file.write(fruit + "\n")

Suppose user enters:

Apple

Python writes:

Apple

into the file.

Run again:

Input:

Banana

File becomes:

Apple
Banana

because append mode adds.

4. READING A FILE
Method 1 — Read Everything
with open("fruits.txt", "r") as file:
    content = file.read()

print(content)

Output:

Apple
Banana
Orange
What is content?

Suppose file contains:

Apple
Banana

Python stores:

content = "Apple\nBanana\n"

inside the variable.

5. READING LINE BY LINE

Instead of reading all at once:

file.read()

Read one line at a time.

Using a Loop
with open("fruits.txt", "r") as file:
    for line in file:
        print(line)

Python does:

line = "Apple\n"
print(line)

line = "Banana\n"
print(line)

and so on.

Why Were There Blank Lines?

Remember this?

print(line)

Output looked like:

Apple

Banana

Orange

because:

line

already contains:

"\n"

and print() adds another one.

FIX
print(line.strip())

Output:

Apple
Banana
Orange
What Does strip() Do?

Removes:

"\n"

and extra spaces.

Example:

"Apple\n"

becomes:

"Apple"

6. COUNTING LINES

Imagine:

John
Mike
Sarah

We want:

Total students: 3

Code:

count = 0

with open("students.txt", "r") as file:
    for line in file:
        count += 1

print(count)
7. SEARCHING INSIDE FILES

File:

Apple
Banana
Orange

Code:

search = input("Search fruit: ")

with open("fruits.txt", "r") as file:
    for line in file:
        if search == line.strip():
            print("Found")

Input:

Banana

Output:
found
