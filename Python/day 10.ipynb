PART 1 — What Is a Nested Loop?

A nested loop means:

A loop inside another loop.

Example:

for i in range(3):
    for j in range(2):
        print(i, j)

Output:

0 0
0 1
1 0
1 1
2 0
2 1

Think of it like:

Outer loop = rows
Inner loop = columns

The inside loop finishes completely before the outside loop moves again.

```python
for i in range(3):
    for j in range(2):
        print(i,j)
        
```

EXERCISE 4 — Triangle Pattern
for i in range(5):
    for j in range(i + 1):
        print("*", end="")
    print()

Output:

*
**
***
****
*****

UNDERSTAND THIS CAREFULLY

Outer loop:

for i in range(5):

Runs:

0
1
2
3
4

Inner loop:

range(i + 1)

Changes every time.

When:

i = 0 → print 1 star
i = 1 → print 2 stars
i = 2 → print 3 stars

That’s the logic.

```python
for i in range(5):
    for j in range(i +1):
        print("*", end= "")
    print()
```

EXERCISE 7 — Count Even Numbers

```python
numbers = [1,2,3,4,5,6]
count = 0
for num in numbers:
    if num %2 == 0:
        count = count+1
print(count)
```

EXERCISE 8 — Find Largest Number
numbers = [4, 8, 2, 15, 7]

largest = numbers[0]

for num in numbers:
    if num > largest:
        largest = num

print(largest)

Output:

15

```python
numbers = [4, 8, 2, 15, 7]
largest = numbers[0]
for num in numbers:
    if num > largest:
        largest = num
print(largest)
```
