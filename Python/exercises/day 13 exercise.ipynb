PART 5 — FINDING VALUES

Problem:

Find largest number.

Think first.

Step 1

```python
def find_largest(numbers):
    largest = numbers[0]
    for num in numbers:
        if num > largest:
            largest = num
    return largest
print(find_largest([2,6,10,7,20]))
```

PART 6 — FINDING SMALLEST

Same idea.

```python
def find_smallest(numbers):
    smallest = numbers[0]
    for num in numbers:
        if num < smallest:
            smallest = num
    return smallest
print(find_smallest([30,10,4,9,3]))
```

PART 7 — COUNTING

A very common Data Analyst task.

Example:

Count even numbers.

```python
def count_even(numbers):
    count = 0
    for num in numbers:
        if num  % 2 == 0:
            count += 1
    return count
print(count_even([2,4,5,7,9]))
```

PART 8 — COUNT ODD NUMBERS

```python
def count_odd(numbers):
    count = 0
    for num in numbers:
        if num % 2 != 0:
            count += 1
    return count
print(count_odd([2,3,24,4,1,6,8,9,0]))
```

PART 9 — SEARCHING

Search for a value.

```python
def find_number(numbers, target):
    for num in numbers:
        if num == target:
            return True
    return False
print(find_number([1,2,3,4,], 9))
```

Search For Names

```python
def find_name(names, target):
    for name in names:
        if name == target:
            return True
    return False
print(find_name(["One", "Nine", "Eight"], "Two"))
```

ART 10 — COUNT OCCURRENCES

Count how many times something appears.

```python
def count_occurrences(numbers, target):
    count = 0
    for num in numbers:
        if num == target:
            count += 1
    return count
print(count_occurrences([1,2,3,2,2,4],2))
```

PART 12 — FILTER EVEN NUMBERS

```python
def filter_even(numbers):
    result = []
    for num in numbers:
        if num %2 == 0:
            result.append(num)
    return result
print(filter_even([1,2,3,4,5,6]))
```

PART 13 — AVERAGE

Very important for data analysis.

Formula:

ext{Average}=rac{ ext{Sum of Numbers}}{ ext{Count of Numbers}}

```python
def calculate_average(numbers):
    total = 0
    for num in numbers:
        total += num
    average = total / len(numbers)
    return average
print(calculate_average([30,50,20,10]))
```

PART 14 — COUNT VOWELS

```python
def count_vowels(word):
    count = 0
    vowels = "aeiou"
    for letter in word:
        if letter in vowels:
            count += 1
    return count
print(count_vowels("education"))
```

PART 15 — LONGEST WORD

```python
def longest_word(words):
    longest = words[0]

    for word in words:
        if len(word) > len(longest):
            longest = word
    return longest
print(longest_word(["cat", "Buffalo", "Elephant"]))
```
