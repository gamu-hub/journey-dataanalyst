Exercise 4 Goal

Ask the user for a movie name.

Check whether that movie exists in movies.txt.

```python
search = input("Search movie:")
found = False

with open("movies.txt", "r") as file:
    for line in file:
        if search == line.strip():
            found = True
if found:
    print("Found!")
else:
    print("Not Found")
```
