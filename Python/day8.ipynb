Part 2 — Loops with Lists (40 min)

This is the BIG skill for today.

numbers = [10, 20, 30, 40]

for num in numbers:
    print(num)

Understand this deeply:

numbers = whole list
num = one item at a time

Now try:

names = ["John", "David", "Sara"]

for name in names:
    print("Hello", name)

```python
numbers = [10, 20, 30, 40]
for num in numbers:
    print(num)
```

```python
names = ["John", "David", "Sara"]
for name in names:
    print("Hello", name)
```

```python
fruits = ["Apple", "Banana", "Orange"]
for fruit in fruits:
    print("I love", fruit)
```
