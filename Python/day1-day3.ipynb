```python
# Task 2 — PUBG Rank

# Ask user rank.

# If:

# Ace → "Pro player"
# Crown → "Good player"
# else → "Keep grinding"

rank = input("Enter your rank!")

if rank == "Ace":
    print("Pro player")

if rank <= "Crown":
    print("Good player")
else:
    print("Keep grinding")
```

```python
# Task 3 — Positive or Negative

# Ask user for a number.

# Check:

# positive
# negative
# or zero

number = int(input("Enter your number"))

if number > 0:
    print("Positive number")
elif number <0:
    print("Negative number")
else:
    print("Zero")
```
