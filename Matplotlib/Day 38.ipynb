```python
import pandas as pd

import matplotlib.pyplot as plt

data= {
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "Sales": [120,150,170,160,210,250],
    "Expenses": [80, 90, 110, 100, 130, 150]
}

df = pd.DataFrame(data)

plt.figure(figsize= (8,4))

plt.plot(
    df["Month"],
    df["Sales"],
    color= "red",
    linestyle= ":",
    marker = "d",
    linewidth= 3,
    markersize= 10
    
)

plt.grid(
    color="grey",
    linestyle= "--"
    )

plt.title("Monthly Sales")
plt.show()
```

```python
plt.figure(figsize=(10,5))

plt.plot(
    df["Month"],
    df["Sales"],
    color="blue",
    linestyle="-",
    marker="o",
    linewidth=3,
    markersize=8,
    label="Sales"
)

plt.plot(
    df["Month"],
    df["Expenses"],
    color="red",
    linestyle="--",
    marker="s",
    linewidth=3,
    markersize=8,
    label="Expenses"
)

plt.title("Monthly Sales vs Expenses")

plt.xlabel("Month")

plt.ylabel("Amount ($)")

plt.grid(True)

plt.legend()

plt.xticks(rotation=45)

plt.show()
```

Bonus Challenge

Using this dataset:

```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "Sales": [120, 150, 170, 160, 210, 250],
    "Expenses": [80, 90, 110, 100, 130, 150]
}

df = pd.DataFrame(data)

plt.figure(figsize= (12,6))

plt.title("Company Performance (First Half of 2026)")

plt.plot(
    df["Month"],
    df["Sales"],
    color= "green",
    marker = "o",
    linewidth = 3,
    markersize = 8,
    label = "Sales"
)

plt.plot(
    df["Month"],
    df["Expenses"],
    color = "orange",
    linestyle = "--",
    marker = "s",
    linewidth = 3,
    markersize = 8,
    label = "Expenses"
)

plt.grid(True)

plt.xlabel("Month")

plt.ylabel("Amount ($)")

plt.legend(loc = "upper left")

plt.xticks(rotation = 45)

plt.show()
```
