```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "Salary":[3000,3200,3400,3500,3600,
              3700,3800,3900,4000,4100,
              4200,4300,4400,4500,9000]
}

df = pd.DataFrame(data)
plt.figure(figsize= (6,4))

plt.boxplot(
    df["Salary"],
    patch_artist=True
    )

plt.figure(figsize=(7,2))

plt.boxplot(
    df["Salary"],
    vert=False
)

plt.title("Employees Salaries")
plt.ylabel("Salary")

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "Salary":[3000,3200,3400,3500,3600,
              3700,3800,3900,4000,4100,
              4200,4300,4400,4500,9000]
}

df = pd.DataFrame(data)

plt.figure(figsize=(6,4))

plt.boxplot(
    df["Salary"],
    patch_artist= True
)

plt.title("Employess Salaries")
plt.ylabel("Salary")
plt.grid(axis= "y")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "Male":[3000,3500,4000,4500,5000],
    "Female":[2800,3300,3900,4200,4600]
}

df = pd.DataFrame(data)

plt.figure(figsize=(6,4))

plt.boxplot(
    [df["Male"], df["Female"]],
    label= ["Male", "Female"]
)


plt.ylabel("Salary")
plt.grid(axis =("y"))
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "Math": [70, 72, 75, 78, 80, 82, 85],
    "Science": [65, 67,70, 72, 75, 95, 98]
}

df= pd.DataFrame(data)

plt.figure(figsize= (6, 4))

plt.boxplot(
    [df["Math"], df["Science"]],
    tick_labels= ["Math", "Science"]
)

plt.title("Math vs Science Scores")
plt.ylabel("Scores")

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "Scores": [65,68,70,72,74,
               75,76,78,80,82,
               84,86,90,92,150]
}

df = pd.DataFrame(data)

plt.figure(figsize=(7,3))

plt.boxplot(
    df["Scores"],
    patch_artist=True,
    vert=False
)

plt.title("Student Scores")
plt.xlabel("Scores")

plt.show()
```
