Goal

Learn how to display multiple charts in one figure. This is very common in dashboards and reports.

What you'll learn
plt.subplot()
plt.subplots()
Multiple charts in one window
Adding titles to each subplot
Using tight_layout()

```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "Month": ["Jan", "Feb", "Mar", "Apr"],
    "Sales": [120, 150, 180, 170],
    "Profit": [20, 30, 40, 35]
}

df = pd.DataFrame(data)

plt.figure(figsize=(10, 4))

plt.subplot(1,2,1)
plt.plot(df["Month"], df["Sales"], marker= "o")
plt.title("Monthly Sales")

plt.subplot(1,2,2)
plt.bar(df["Month"], df["Profit"])
plt.title("Monthly Profit")

plt.tight_layout()
plt.show()
```

Understanding subplot(rows, columns, position)
plt.subplot(2,2,1)

means:

2 rows
2 columns
First position

Layout:

+-----+-----+
|  1  |  2  |
+-----+-----+
|  3  |  4  |
+-----+-----+

Example 2: Four Charts

```python
plt.figure(figsize=(10,4))

plt.subplot(1,2,1)
plt.plot(df["Month"], df["Sales"], marker="o")
plt.title("Monthly Sales")

plt.subplot(1,2,2)
plt.bar(df["Month"], df["Profit"])
plt.title("Monthly Profit")

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "Month": ["Jan","Feb","Mar","Apr"],
    "Sales": [120,150,180,170],
    "Profit": [20,30,40,35]
}

df = pd.DataFrame(data)

plt.figure(figsize=(10,8))

plt.subplot(2,2,1)
plt.plot(df["Month"], df["Sales"], marker="o")
plt.title("Line Chart")

plt.subplot(2,2,2)
plt.bar(df["Month"], df["Sales"])
plt.title("Bar Chart")

plt.subplot(2,2,3)
plt.scatter(df["Sales"], df["Profit"])
plt.title("Scatter Plot")

plt.subplot(2,2,4)
plt.pie(
    df["Sales"],
    labels=df["Month"],
    autopct="%1.1f%%"
)
plt.title("Pie Chart")

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "Product": ["A", "B", "C", "D"],
    "Sales": [150, 200, 180, 220],
    "Profit": [30, 50, 40, 60]
}
#Create a 2×2 subplot containing:

# Line chart of Sales
# Bar chart of Sales
# Scatter plot (Sales vs Profit)
# Pie chart of Sales

df= pd.DataFrame(data)

plt.figure(figsize= (8,8))
plt.subplot(2,2,1)
plt.plot(df["Product"], df["Sales"], marker= "o")
plt.title("Line Chart")

plt.subplot(2,2,2)
plt.bar(df["Product"], df["Sales"])
plt.title("Bar Chart")

plt.subplot(2,2,3)
plt.scatter(df["Sales"], df["Profit"])
plt.title("Scatter plot")

plt.subplot(2,2,4)
plt.pie(
    df["Sales"],
    labels=df["Product"],
    autopct="%1.1f%%"
)

plt.title("Pie Chart")

plt.tight_layout()

plt.show()
```
