# Exercise 2
Write a Python program that:

1. Asks the user to input a sentence.

2. Checks if the sentence is made up of only alphabetic characters. If not, print how many non-alphabetic characters are in the sentence.

3. Determines if the sentence ends with an exclamation mark (!).

4. Finds out if the sentence contains any whitespace characters and prints a message accordingly.

```python
# Asks the users to input a sentence
sentence = input("Write your sentence")

# Check if the sentence contains only alphabetic characters
if sentence.isalpha():
    print("The sentence is made up of only alphabetic characters")
else:
    non_alpha = 0
    for char in sentence:
        if not char.isalpha():
            non_alpha += 1

    print("Number of non-alphabetic characters:", non_alpha)

# Check if the sentence ends with !

if sentence.endswith("!"):
    print("The sentence ends with an exclamation mark.")

else:
    print("The sentence does not end with an exclamation mark.")

# Find if the sentence contains any whitespace characters

had_whitespace = False

for char in sentence:
    if char.isspace():
        had_whitespace = True
        break
if had_whitespace:
    print("The sentence contains whitespace")
else:
    print("The sentence does not contains whitespace")
```
