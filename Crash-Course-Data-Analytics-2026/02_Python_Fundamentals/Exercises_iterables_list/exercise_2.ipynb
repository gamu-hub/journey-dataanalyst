Using the same list from the previous exercise:

1. Add "fig" to the end of the fruits

2. Insert "grape" at the beginning of the list

3. Remove "cherry" from the list using using the specific method for it

4. Remove the last element from the list using the specific method for it

5. Create another list of berries and combine it with the fruits list into a list called "combined_list"

6. Sort the combined_list

7. Slice the last three elements of the combined list

```python
fruits = ["apple", "banana", "cherry", "date", "elderberry"]

# add fig to the end of the fruits

fruits.append("fig")
print(fruits)

fruits.insert(0, "grape")
print(fruits)

fruits.remove("cherry")
print(fruits)
print("\nRemove the last element from the list using the specific method for it")

fruits.pop()
print(fruits)

# Create another list of berries and combine it with the fruits list into a list called "combined_list"

berries_list = ["strawberry", "blueberries", "greenberries"]
combine_list = berries_list + fruits
print(combine_list)

combine_list.sort()
print(combine_list)

print(combine_list[-3:])
```
