```python
numbers = [1, 2, 3, 4, 5]

for number in numbers:
    print(number)

# we can also print in reverse

print("\n")

numbers = [1,2,3,4,5]

for number in reversed(numbers):
    print(number)
```

# we see the difference in the print output. 

We can use "\n" to make space between two results but end= "" can be use in different wat like above. 

```python
numbers = [1, 2, 3, 4, 5]

for number in numbers:
    print(number, end="")

# we can also print in reverse

print("\n")

for number in numbers:
    print(number, end= "")


```

```python
# One new thing is end= "". Instead we can add end= "_"

numbers = [1,2,3,4,5]

for number in reversed(numbers):
    print(number, end= "-")
```

```python
# sets can't be reversible

fruits = {"apple", "orange", "banana", "coconut"}

for fruit in reversed (fruits):
    print(fruit)
```

```python
my_dictionary = {"A": 1, "B": 2, "C": 3}

for key, value in my_dictionary.items():
    print(key, value)

print("\nusing the f-string")

for key, value in my_dictionary.items():
    print(f"{key} = {value}")
```
