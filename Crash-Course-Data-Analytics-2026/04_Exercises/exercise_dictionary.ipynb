```python
# pairs = [("names", "Ellie"), ("job", "Instructor")]

# {
#     "name": "Ellie",
#     "job": "Instructor"
# }

pairs = [("name", "Elie"), ("job", "Instructor")]

result = {}

for pair in pairs:
    key = pair[0]
    value = pair[1]
    result[key] = value
print(result)
```

Super Bonus:
Given the string "awesome sauce", return a dictionary where the keys are vowels, and the values are the count of each vowel in the string. Your dictionary should look like this: {'a': 2, 'e': 3, 'i': 0, 'o': 1, 'u': 1}.



```python
# text' = "awesome sauce"

# {
#     'a' : 2,
#     'e' : 3,
#     'i' : 0,
#     'o' : 1,
#     'u': 1
# }

text = "awesome sauce"
vowels = "aeiou"

result = {}

for vowel in vowels:
    result[vowel] = 0

for letter in text:
    if letter in vowels:
        result[letter] += 1
print(result)
```
