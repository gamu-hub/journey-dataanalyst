1. What is a CSV File?

CSV = Comma Separated Values

It stores table data like Excel.

Example:

Name,Age,Country
John,25,USA
David,30,Israel
Sara,22,India

Each row = one record
Each column = one type of information

Reading a CSV File

Use:

pd.read_csv()

Example:

import pandas as pd

df = pd.read_csv("students.csv")

print(df)

.head()

Shows the first 5 rows

print(df.head())

You can also choose number of rows:

print(df.head(3))

Useful when datasets are very large.

5. .tail()

Shows the last 5 rows

print(df.tail())

Custom rows:

print(df.tail(2))

6. .shape

Shows:

(rows, columns)

Example:

print(df.shape)

Output:

(100, 4)

Meaning:

100 rows
4 columns

7. .columns

Shows all column names.

print(df.columns)

Output:

Index(['Name', 'Age', 'Country'], dtype='object')

Full Example
students.csv
Name,Age,Country
John,25,USA
David,30,Israel
Sara,22,India
Mike,28,Canada
Anna,24,Korea





```python
import pandas as pd
df = pd.read_csv("students.csv")

print("FIRST ROWS")
print(df.head())

print("\nLAST ROWS")
print(df.tail())

print("\nSHAPE")
print(df.shape)

print("\nCOLUMNS")
print(df.columns)
```

full code for what we learn today.

```python
import pandas as pd

data = {
    "Team": ["Team HOPE", "ELV esports", "Muvanlai esports", "PLE esports"],
    "Points": [80, 70, 60, 75],
    "Country": ["Israel", "USA", "India", "Korea"]
}

df = pd.DataFrame(data)

print("ONE COLUMN")
print(df["Team"])

print("\nMULTIPLE COLUMNS")
print(df[["Team", "Points"]])

print("\nFIRST ROW")
print(df.iloc[0])

print("\nFILTER POINTS > 70")
print(df[df["Points"] > 70])

print("\nSORT BY POINTS")
print(df.sort_values(by="Points", ascending=False))
```

Mini Exercise

Using the same dataset:

Print only "Country" column
Print rows 1 to 3 using iloc
Find teams with points less than 75
Find teams from "USA"
Sort by "Team" alphabetically

```python
import pandas as pd
data = {
    "Team": ["Team HOPE", "ELV esports", "Muvanlai esports", "PLE esports"],
    "Points": [80, 70, 60, 75],
    "Country": ["Israel", "USA", "India", "Korea"]
}

df = pd.DataFrame(data)

print("COUNTRY")
print(df["Country"])

print("\nFIRST THREE ROWS")
print(df.iloc[0:3])

print("\nTEAM LESS THAN 75")
print(df[df["Points"] < 75])

print("\nFind teams")
print(df[df["Country"] == "USA"])

print("\nSORTED TEAMS")
print(df.sort_values(by= "Team",))
```
