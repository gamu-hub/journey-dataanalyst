9. Filtering Rows (Real Analyst Thinking)

Now the fun starts.

Example:

Show only people older than 24.

print(df[df["Age"] > 24])

You will see we use [df][df] the reason is below.

We use df two times because the inner df["Age"] > 24 creates the True/False condition, and the outer df[...] uses that condition to filter the rows.

```python
import pandas as pd
data = {
    "Name": ["Lootpu", "Niang", "Lotzempa"],
    "Age": [35, 30, 25],
    "City": ["Nakhalim", "Kiryat Arba", "Tel Aviv"]
}

df = pd.DataFrame(data)
print(df[df["Age"] > 27])
```

Filter by Text
print(df[df["City"] == "Tel Aviv"])

```python
import pandas as pd
data = {
    "Name": ["Lootpu", "Niang", "Lotzempa"],
    "Age": [35, 30, 25],
    "City": ["Nakhalim", "Kiryat Arba", "Tel Aviv"]
}

df = pd.DataFrame(data)
print(df[df["City"]== "Tel Aviv"])
```
