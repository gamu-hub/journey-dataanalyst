Step 4: Cleaning the Data
🤔 Why do we clean data?

Imagine your manager asks:

"What is the average selling price of our products?"

We must use 

average_price = df["Price"].mean()
print("Average Price:", average_price)

```python
import pandas as pd

data = {
    "Product":[
        "Laptop", "Mouse", "Keyboard", "Laptop",
        "Monitor", None, "Mouse", "Keyboard"
    ],
    "Price": [
        1200, 25, 70, None,
        300, 50, 25, 70
    ],
    "Quantity":[
        5, 20, None, 3,
        4, 10, None, 8
    ],
    "Country": [
        "USA", "Canada", "USA", None,
        "Germany", "Canada", "USA", None
    ]
}

df = pd.DataFrame(data)

# Check the Price column before cleaning
print(df["Price"])

# Calculate the average (mean) price
average_price = df["Price"].mean()
print("Average Price:", average_price)
```

Suppose the manager says:

"Now clean the dataset because we'll use it for future analysis."

One row has:

Product	Price
Laptop	NaN

If we leave it like that, future calculations (like Sales = Price × Quantity) may be incomplete or harder to work with.

So we replace the missing value.

df["Price"] = df["Price"].fillna(average_price)

This means:

"Take every missing price and replace it with the average price."

```python
import pandas as pd

data = {
    "Product":[
        "Laptop", "Mouse", "Keyboard", "Laptop",
        "Monitor", None, "Mouse", "Keyboard"
    ],
    "Price": [
        1200, 25, 70, None,
        300, 50, 25, 70
    ],
    "Quantity":[
        5, 20, None, 3,
        4, 10, None, 8
    ],
    "Country": [
        "USA", "Canada", "USA", None,
        "Germany", "Canada", "USA", None
    ]
}

df = pd.DataFrame(data)


# ==============================
# Clean the Price Column
# ==============================

# Step 1: Check the Price column before cleaning
print("Price column before cleaning:")
print(df["Price"])

# Step 2: Calculate the average price
average_price = df["Price"].mean()

# Step 3: Display the average price
print("\nAverage Price:", average_price)

# Step 4: Replace missing (NaN) values with the average price
df["Price"] = df["Price"].fillna(average_price)

# Step 5: Check the Price column after cleaning
print("\nPrice column after cleaning:")
print(df["Price"])

```
