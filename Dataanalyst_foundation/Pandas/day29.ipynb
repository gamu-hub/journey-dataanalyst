What is NaN?

NaN means:

Not a Number

It represents missing or empty values inside a DataFrame.

Example:

Team	Kills
HOPE	45
ELV	NaN
PLE	38

Here,

ELV's kills are missing.

```python
import pandas as pd

data = {
    "Team": ["HOPE", "ELV", "PLE", "Muvanlai"],
    "Kills": [45, None, 38, 50],
    "Country": ["Israel", "USA", None, "India"]
}

df = pd.DataFrame(data)

print(df)
```

Fill Missing Values

Instead of deleting rows, we can replace NaN.

Example:

df["Kills"] = df["Kills"].fillna(0)
print(df)

```python
import pandas as pd

data = {
    "Team": ["HOPE", "ELV", "PLE", "Muvanlai"],
    "Kills": [45, None, 38, 50],
    "Country": ["Israel", "USA", None, "India"]
}

df = pd.DataFrame(data)
df["Kills"] = df["Kills"].fillna(0)
print(df)
```

Task 6

Print the cleaned DataFrame.

```python
import pandas as pd
data = {
    "Player" : ["Alex", "John", "Sarah", "Mike", "Emma", "David"],
    "Kills": [45, None, 52, 38, None, 41],
    "Matches": [5, 4, None, 6, 5, None],
    "Country": ["USA", "Canada", None, "USA", "Korea", None]
}

df = pd.DataFrame(data)

print("Count missing values")
print(df.isnull().sum())

print("\nRows containing at least one missing value")
print(df[df.isnull().any(axis=1)])

print("\nFill missing values in Kills with the mean")
df["Kills"]=df["Kills"].fillna(df["Kills"].mean())
print(df)

print("\nFill missing values in Matches with the median")
df["Matches"] = df["Matches"].fillna(df["Matches"].median())
print(df)

print("\nReplacing missing values in the Country column with Unknown")
df["Country"] = df["Country"]. fillna("Unknown")
print(df)

print("\nCleaned DataFrame")
print(df)
```

One thing to remember

For practice, it's a good idea to print after every step because you can see exactly what changed.

For real-world work, analysts usually write:

```python
df["Kills"] = df["Kills"].fillna(df["Kills"].mean())
df["Matches"] = df["Matches"].fillna(df["Matches"].median())
df["Country"] = df["Country"].fillna("Unknown")

print(df)
```
