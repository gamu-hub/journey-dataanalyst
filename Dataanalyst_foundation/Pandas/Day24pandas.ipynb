1. Create a Simple DataFrame

```python
import pandas as pd

data = {
    "Team": ["Team HOPE", "ELV esports", "Muvanlai esports", "PLE esports"],
    "Kills": [40, 35, 28, 50],
    "Matches": [5, 5, 5, 5]
}

df = pd.DataFrame(data)

print(df)
```

5. Update Specific Rows

Use .loc[]

Example: Change Team HOPE kills to 60

print(df)
Meaning
loc[row, column]

So:

df.loc[0, "Kills"]

```python
import pandas as pd

data = {
    "Team": ["Team HOPE", "ELV esports", "Muvanlai esports", "PLE esports"],
    "Kills": [40, 35, 28, 50],
    "Matches": [5, 5, 5, 5]
}

df = pd.DataFrame(data)

# Add average kills
df["Avg Kills"] = df["Kills"] / df["Matches"]

# Add rank points
df["Points"] = df["Kills"] * 2

# Strong team check
df["Strong Team"] = df["Kills"] > 40

# Update one value
df.loc[1, "Kills"] = 45

print(df)
```

```python
import pandas as pd

data = {
    "Team": ["Team HOPE", "ELV esports", "Muvanlai esports", "PLE esports"],
    "Kills": [40, 35, 28, 50],
    "Matches": [5, 5, 5, 5]
}

df = pd.DataFrame(data)

df["Avg Kills"] = df["Kills"] / df["Matches"]

df["Tournament"] = "EIMI CUP"

df["Kills"] = df["Kills"] + 5

df.loc[0, "kills"] = 60

df["Frag team"] = df["Kills"] > 40

df["Bonus"] = 0
df.loc[df["Kills"] > 40, "Bonus"] = 10

df.rename(columns={"Kills": "Total Kills"}, inplace=True)

df.drop(columns=["Bonus"], inplace=True)


print(df)
```

Exercise 5

Now let’s combine calculations + conditions together.

Task

Create a new column called:

"Points"

Formula:

(Kills * 2) + Matches

Then create another column:

"Rank"

Rules:

If Points > 80 → "Top Tier"
Otherwise → "Mid Tier"

```python
import pandas as pd

data = {
    "Team": ["Team HOPE", "ELV esports", "Muvanlai esports", "PLE esports"],
    "Kills": [40, 35, 28, 50],
    "Matches": [5, 5, 5, 5]
}

df = pd.DataFrame(data)
df["Points"] = (df["Kills"] *2) + df["Matches"]

df["Ranks"] = "Mid Tier"
df.loc[df["Points"] > 80, "Ranks"] = "Top Tier"

print(df)
```
