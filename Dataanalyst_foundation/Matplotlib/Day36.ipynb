Step 1: Basic Pie Chart

```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "Department": ["Sales", "Marketing", "IT", "HR"],
    "Employees": [40, 25, 20, 15]
}

df = pd.DataFrame(data)

plt.figure(figsize=(6,6))

plt.pie(df["Employees"])

plt.title("Company Employees")

plt.show()
```

Step 2: Add Labels
now evert slice has its department name. 

```python
plt.figure(figsize=(6,6))

plt.pie(
    df["Employees"],
    labels=df["Department"]
)

plt.title("Company Employees")

plt.show()
```

Step 3: Show Percentages



```python
plt.figure(figsize=(6,6))

plt.pie(
    df["Employees"],
    labels=df["Department"],
    autopct="%1.1f%%"
)

plt.title("Company Employees")

plt.show()
```

What does this mean?
autopct="%1.1f%%"

Breakdown:

% → formatting
1 → minimum width
.1 → one decimal place
f → floating-point number
%% → display a literal % sign

Step 4: Add Colors

```python
colors = ["skyblue", "orange", "green", "red"]

plt.figure(figsize=(6,6))

plt.pie(
    df["Employees"],
    labels=df["Department"],
    autopct="%1.1f%%",
    colors=colors
)

plt.title("Company Employees")

plt.show()
```

Step 5: Rotate the Pie

```python
plt.figure(figsize=(6,6))

plt.pie(
    df["Employees"],
    labels=df["Department"],
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Company Employees")

plt.show()
```

Step 6: Highlight a Slice (Explode)
explode = [0, 0, 0.1, 0]

```python
plt.figure(figsize=(6,6))

plt.pie(
    df["Employees"],
    labels=df["Department"],
    autopct="%1.1f%%",
    explode=explode
)

plt.title("Company Employees")

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "Department": ["Sales", "Marketing", "IT", "HR"],
    "Employees": [40, 25, 20, 15]
}

df = pd.DataFrame(data)

plt.figure(figsize=(6,6))

plt.pie(
    df["Employees"],
    labels=df["Department"],
    autopct="%1.1f%%",
    startangle=90,
    colors= ["skyblue", "purple", "green", "red"],
    explode= [0,0.2,0,0]
    )

plt.title("Employee Distribution")

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "Fruit": ["Apple", "Banana", "Orange", "Mango"],
    "Sales": [50, 30, 15, 5]
}

df = pd.DataFrame(data)

plt.pie(
    df["Sales"],
    labels=df["Fruit"],
    autopct="%1.1f%%",
    colors= ["green", "yellow", "blue", "purple"],
    startangle=90,
    explode= [0.2, 0, 0, 0]

)

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

data= {
    "Category": ["Rent", "Food", "Transport", "Entertainment", "Saving"],
    "Amount": [ 1200, 450, 150, 200, 500]
}

df = pd.DataFrame(data)

plt.figure(figsize= (7,7))

plt.pie(
    df["Amount"],
    labels= df["Category"],
    autopct="%1.1f%%",
    colors= ["gold", "lightgreen", "skyblue", "tomato", "violet"],
    startangle= 90,
    explode= [0,0,0,0,0.3]

)

plt.title("Monthly Expenses Distribution")

plt.show()
```
