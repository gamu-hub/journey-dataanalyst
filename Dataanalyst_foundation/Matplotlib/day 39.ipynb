```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "Month": ["Jan","Feb","Mar","Apr","May","Jun"],
    "Sales": [120,150,170,160,210,250],
    "Expenses": [80,90,110,100,130,150]
}

df = pd.DataFrame(data)

plt.figure(
    figsize=(10,5),
    facecolor= "lightyellow")

plt.plot(
    df["Month"],
    df["Sales"],
    color = "green",
    linestyle= "--",
    linewidth= 4,
    marker = "o",
    markersize = 12,
    markerfacecolor= "yellow",
    markeredgecolor= "black",
    label = "Sales"
    
)

plt.plot(
    df["Month"],
    df["Expenses"],
    color = "red",
    linestyle= "--",
    linewidth = 2,
    marker = "s",
    label = "Expenses"
)

plt.title("Monthly Sales vs Expenses")
plt.xlabel("Month")
plt.ylabel("Amount")

plt.grid(
    color = "grey",
    linestyle = "--",
    linewidth= 0.5
)
plt.xlim("Feb", "Jun")
plt.ylim(100, 260)

plt.legend(loc= "upper left")

plt.savefig("Sales_chart.png",dpi= 300)

plt.show()


```
