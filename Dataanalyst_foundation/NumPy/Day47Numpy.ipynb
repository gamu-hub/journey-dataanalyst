Part 3 – Selecting Multiple Columns

```python

import numpy as np
scores = np.array([
    [80,90,70],
    [60,75,85],
    [95,88,92],
    [50,65,70]
])

print(scores[:,[0,2]])

# Meaning

# All rows

# Columns 0 and 2
```

Part 4 – Reordering Rows

Fancy indexing also changes order.

```python
import numpy as np
scores = np.array([
    [80,90,70],
    [60,75,85],
    [95,88,92],
    [50,65,70]
])

print(scores[[2,0,3]])
# Notice the rows appear in the order you requested.
```

Part 8 – Fancy Indexing vs Slicing

Slicing

```python
import numpy as np
numbers = np.array([10,20,30,40,50])


print("Fancy indexing")

print(numbers[[1,4]])

print("\nSlicing Indexing")
print(numbers[1:4])
```

numbers = np.array([10,20,30,40,50])

```python
scores = np.array([
    [80,90,70],
    [60,75,85],
    [95,88,92],
    [50,65,70]
])

print(scores[scores[:, 0] > 70])
```

```python
# Exercise 4 (Mini Challenge)
import numpy as np

sales = np.array([
    [120,450,700],
    [900,250,180],
    [550,610,90],
    [800,420,300]
])
print("Printing only from stores 2 and 4")
print(sales[[1,3]])

print("\nPrint only Feb and Mar")
print(sales[:, [1,2]])

print("\nStores where Jan sales are above 500")
print(sales[sales[:,0] > 500])

print("\nCounting how many stores have jan sales above 500")
print(np.sum(sales[:,0] > 500))

print("\nAverage Jan sales for those stores")
print(np.mean(sales[sales[:,0] > 500, 0]))

# Print only stores:
# 2
# 4

# Print only February and March

# Print stores where January sales are above 500

# Count how many stores have January sales above 500

# Find the average January sales for those stores
```

An easier way to read it

Break it into two steps:

```python
jan_sales = sales[:,0]

high_sales = jan_sales[jan_sales > 500]

print(np.mean(high_sales))
```
