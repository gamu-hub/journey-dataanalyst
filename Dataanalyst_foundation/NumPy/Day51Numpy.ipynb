# 6. Logarithm

Suppose:

10² = 100

We know the answer is 100.

A logarithm asks the reverse question:

"10 needs to be raised to what power to become 100?"

Answer:

2

```python
numbers = np.array([1,10,100,1000])

print(np.log10(numbers))
```

# 7. Rounding Numbers
# np.round(numbers, 2)

means:

Round every number in numbers to 2 decimal places.

np.round(3.14159, 2)

→ 3.14

np.round(3.14159, 3)

→ 3.142

np.round(3.14159, 1)

→ 3.1

# B. np.floor()

floor means:

Always round DOWN to the nearest whole number.

# C. np.ceil()

ceil is the opposite.
numbers = np.array([3.1, 2.2, 9.01])

print(np.ceil(numbers))

Output:

[ 4.  3. 10.]

```python
numbers = np.array([3.14159,2.71828,9.87654])

print(np.round(numbers,2))

print(np.floor(numbers))

print(np.ceil(numbers))
```

```python
# Exercise 2 — Prices

import numpy as np

prices = np.array([12.99, 45.50, 9.99, 100.75, 55.30])

print("Round to 1 decimal")
print(np.round(prices))

print("\nRound up:")
print(np.ceil(prices))

print("\nRound Down:")
print(np.floor(prices))

print("\nCalculate a 15% discount")
discount = prices * 0.15
discounted_prices = prices - discount
print(discounted_prices)

# Round to 1 decimal
# Round up
# Round down

# Calculate a 15% discount
```

🏆 Day 51 — Mini Challenge



```python
import numpy as np

profits = np.array([
    [1200, 1500, 1800],
    [2200, 2400, 2600],
    [900, 1100, 1300]
])

print("Adding 10% to every profit")
increase = profits * 0.10
new_profits = profits + increase
print(new_profits)

print("\nround the result")
print(np.round(new_profits))

print("\nCalculating square root")
print(np.sqrt(new_profits))

print("\nFinding the highest value")
print(np.max(new_profits))
# Task 1 — Add 10% to every profit
# Task 2 — Round the result
# Task 3 — Calculate square root
# Task 4 — Find the highest value
```
