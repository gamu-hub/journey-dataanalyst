3. Transpose

Rows become columns.

````python
import numpy as np

numbers = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

print("Original")
print(numbers)

print("\nTranspose")
print(numbers.T)
````

4. Swap Axes

Useful for 3D arrays.

````python
import numpy as np

array = np.arange(24).reshape(2,3,4)
print(array.shape)

new = array.swapaxes(0, 1)

print(new.shape)
````

```python
**5. Squeeze

Removes dimensions of size 1.**
```

````python
array = np.array([[[10,20,30]]])
print(array.shape)

new = np.squeeze(array)

print(new.shape)
print(new)
````

6. Expand Dimensions
Adding a new axis

````python
import numpy as np

arr = np.array([10,20,30])
print(arr.shape)

new = np.expand_dims(arr, axis= 0)
print(new.shape)

print(new)

new = np.expand_dims(arr, axis=1)
print(new)
````

7. Resize an Array

````python
import numpy as np

arr = np.array([1,2,3,4])

new = np.resize(arr, (2,4))
print(new)
````

🎯 Mini Exercise

scores = np.array([
    [75, 80, 90],
    [60, 70, 85],
    [95, 88, 92]
])

Complete these tasks:

* Print the original array.
* Flatten it.
* Use ravel().
* Print the transpose.
* Print the shape.
* Convert it into a column vector using reshape(-1, 1).
* Convert it into a row vector using reshape(1, -1).
* Use expand_dims() with axis=0.
* Use expand_dims() with axis=1.
* Print the shape after every operation.

````python
import numpy as np

scores = np.array([
    [75, 80, 90],
    [60, 70, 85],
    [95, 88, 92]
])

print("Original array")
print(scores)

flat = scores.flatten()
print("\nFlattened:")
print(flat)

print("\nUsing ravel()")
print(scores.ravel())

print("\nPrinting the transpose")
print(scores.T)

print("\nPrinting shape")
print(scores.shape)

print("\nConverting Column vector using reshape")
column = scores.reshape(-1,1)

print(column)
print(column.shape)

print("\nConverting Row vector using reshape")
row = scores.reshape(1, -1)

print(row)
print(row.shape)

print("\nUsing expand_dims() with axis=0.")
new = np.expand_dims(scores, axis= 0)
print(new.shape)

print("\nUsing expand_dims() with axis=1.")
new = np.expand_dims(scores, axis=1)
print(new.shape)
````
