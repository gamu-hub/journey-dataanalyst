```python
# 📝 Exercise 3

import numpy as np

prices = np.array([
    [120, 450, 700],
    [900, 250, 180],
    [550, 610, 90]
])

# Replace every value below 200 with 200
# Print all values above 500
# Count how many values are above 500
# Find the average of values above 500


print("\nReplacing value below 200 with 200")
prices[prices < 200] = 200
print(prices)

print("\nAll values above 500")
print(prices[prices > 500])

print("\nCounting values above 500:")
print(np.sum(prices > 500))

print("\nAverage values above 500")
print(np.mean(prices[prices > 500]))
```
