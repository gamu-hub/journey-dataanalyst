
# 1. Basic Indexing

```python
import numpy as np

sales = np.array([
    [120, 150, 180, 200],
    [300, 280, 320, 350],
    [500, 450, 470, 520],
    [250, 270, 260, 290]
])

print(sales[0,0])
```

# 2. Slicing

Slicing lets you select multiple values.

```python
import numpy as np

sales = np.array([
    [120, 150, 180, 200],
    [300, 280, 320, 350],
    [500, 450, 470, 520],
    [250, 270, 260, 290]
])

print(sales[0:2])
```

Selecting specific rows and columns

```python
import numpy as np

sales = np.array([
    [120, 150, 180, 200],
    [300, 280, 320, 350],
    [500, 450, 470, 520],
    [250, 270, 260, 290]
])
print(sales[1:3, 0:2])
```

3. Slicing with :

A colon by itself means everything.

```python
import numpy as np

sales = np.array([
    [120, 150, 180, 200],
    [300, 280, 320, 350],
    [500, 450, 470, 520],
    [250, 270, 260, 290]
])
print(sales[:, 1])
print(sales[2, :]) #meaning Row 2, all columns.
```

# 4. Boolean Indexing 
You can ask NumPy:

"Give me only the values that satisfy this condition."



```python
import numpy as np

sales = np.array([
    [120, 150, 180, 200],
    [300, 280, 320, 350],
    [500, 450, 470, 520],
    [250, 270, 260, 290]
])

print(sales[sales > 400])
```

# Multiple conditions

```python
import numpy as np

sales = np.array([
    [120, 150, 180, 200],
    [300, 280, 320, 350],
    [500, 450, 470, 520],
    [250, 270, 260, 290]
])

print(sales[(sales > 250) & (sales < 500)])
#Give me values greater than 250 AND less than 500.
```

# 5. Fancy Indexing

Fancy indexing means selecting specific rows or columns using an array/list of indexes.

```python
import numpy as np

sales = np.array([
    [120, 150, 180, 200],
    [300, 280, 320, 350],
    [500, 450, 470, 520],
    [250, 270, 260, 290]
])

print(sales[[0, 2]])
print(sales[:, [0, 2]]) #You can also select columns:
```

```python
import numpy as np

sales = np.array([
    [120, 150, 180, 200],
    [300, 280, 320, 350],
    [500, 450, 470, 520],
    [250, 270, 260, 290]
])
print("Print the value 470")
print(sales[2,2])

print("\nPrinting entire second row")
print(sales[1])

print("\nPrinting the first three rows")
print(sales[0:3])

print("\nPrinting the last two columns")
print(sales[:, [2,3]])

print("\nAll values greater than 300")
print(sales[sales > 300])

print("\nAll values between 200 and 400")
print(sales[(sales > 200) & (sales < 400)])

print("\nPrinting only rows 1 and 3")
print(sales[[1,3]])

print("\nprinting only columns 0 and 3")
print(sales[:,[0,3]])

print("\nBonus challenge")
new_sales = (sales[:,[2,3]])
print(new_sales[new_sales > 250])
```
