1. Median

The median is the middle value after sorting.

```python
import numpy as np

numbers = np.array([4, 9, 2, 10, 100])

print(np.median(numbers))
```

2. Standard Deviation

Standard deviation tells us how spread out the data is.

Small standard deviation

→ Numbers are close together.

Large standard deviation

→ Numbers are spread apart.

```python
import numpy as np

scores = np.array([70,72,74,76,78])

print(np.std(scores))
```

# 3. Variance

Variance is similar to standard deviation.

```python
import numpy as np

scores = np.array([70,72,74,76,78])

print(np.var(scores))
```

4. Count

Count how many values exist.

```python
import numpy as np

scores = np.array([70,72,74,76,78])

print(scores.size)
```

```python
# Exercise 2

import numpy as np

profits = np.array([
    [1200,1400,1500],
    [1800,1700,1600],
    [2200,2400,2300]
])
print("Average profit of each company")
print(np.mean(profits, axis= 1))

print("\nMedian profit of each company")
print(np.median(profits, axis= 1))

print("\nhighest profit of each company")
print(np.max(profits, axis = 1))

print("\nlowest profit of each company")
print(np.min(profits, axis= 1))
# Print average profit of each company
# Print median profit of each company
# Print highest profit of each company
# Print lowest profit of each company
```

```python
import numpy as np

sales = np.array([
    [120,150,180,200],
    [300,280,320,350],
    [500,450,470,520],
    [250,270,260,290]
])

print("Average sales for each store")
print(np.mean(sales, axis= 1))

print("\nMedian sales for each store")
print(np.median(sales, axis= 1))

print("\nHighest sales for each store")
print(np.max(sales, axis= 1))

print("\nLowest sales for each store")
print(np.min(sales, axis= 1))

print("\nStandard deviation for each store")
print(np.std(sales, axis= 1))

print("\nThis store has the highest average sales")
average_sales = np.mean(sales, axis= 1)
highest_store = np.argmax(average_sales)
print(highest_store)

# 1. Print average sales for each store.
# 2. Print median sales for each store.
# 3. Print the highest sales for each store.
# 4. Print the lowest sales for each store.
# 5. Print the standard deviation for each store.
# 6. Which store has the highest average sales?
```
