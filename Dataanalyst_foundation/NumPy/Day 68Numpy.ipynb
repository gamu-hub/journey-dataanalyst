And we'll keep reinforcing your () vs [] distinction as we go.

```python
import numpy as np

sales = np.array([
    [120, 150, 180, 200],
    [300, 280, 320, 350],
    [500, 450, 470, 520],
    [250, 270, 260, 290]
])

print("Task 1")
result_1 = np.where(sales < 250, 0, sales)
print(result_1)

print("\nTask 2")
result_2 = np.where(sales >= 400, 1, 0)
print(result_2)

print("\nTask 3")
result_3 = np.where(sales >= 400, "High", "Normal")
print(result_3)

print("\nTask 4")
result_4 = np.where(sales > 450)
print(result_4)

print("\nBONUS")
result = np.where(sales > 300, 100, sales)
All sales greater than 300 will be replace by 100 and the rest will be unchange
```
