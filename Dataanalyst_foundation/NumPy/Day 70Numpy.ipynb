1. Basic idea

Suppose we have:

sales = np.array([120, 250, 400, 550, 700])

We want every value to be between 200 and 500.

result = np.clip(sales, 200, 500)


print(result)


```python
import numpy as np

sales = np.array([
    [120, 150, 180, 200],   # Product A
    [300, 280, 320, 350],   # Product B
    [500, 450, 470, 520],   # Product C
    [250, 270, 260, 290]    # Product D
])

print("Task 1")
result_1 = np.clip(sales, 150, 400)
print(result_1)

print("\nTask 2")
result_2 = np.clip(sales, 250, 500)
print(result_2)

print("\nTask 3")
result_3 = np.clip(sales, 100, 300)
print(result_3)

print("\nTask 4")
result_4 = The result will show only sales between 100 and 200
```
