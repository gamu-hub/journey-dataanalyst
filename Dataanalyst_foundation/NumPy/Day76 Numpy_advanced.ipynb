# 1. np.where() — Conditional Logic

You already know how to filter:

```python
import numpy as np

sales = np.array([100, 250, 80, 400, 150])

print(sales[sales > 200])
```

# But sometimes we don't want to filter the array.

We want to change values based on a condition.

That's where np.where() comes in.

Example

```python
sales = np.array([100, 250, 80, 400, 150])

result = np.where(sales > 200, "High", "Low")

print(result)
```

```python
scores = np.array([45, 72, 88, 51, 39])
result = np.where(scores >= 60, "Pass", "Fail")

print(result)
```

# 2. np.select() — Multiple Conditions

What if we have more than two categories?

For example:

Sales < 100 → Low
Sales 100–250 → Medium
Sales > 250 → High

We can use np.select().

```python
import numpy as np
sales = np.array([50, 120, 300, 200, 80, 450])

conditions = [
    sales < 100,
    (sales >= 100) & (sales <= 250),
    sales > 250
]

choices = [
    "Low",
    "Medium",
    "High"
]

result = np.select(conditions, choices, default = "Unknown")

print(result)
```

# 3. np.unique() — Find Unique Values

Suppose we have customer regions:

```python
regions = np.array([
    "North",
    "South",
    "North",
    "East",
    "South",
    "North"
])

print(np.unique(regions))
```

# It removes duplicates.

Count unique values

You can also get how many times each value appears:

```python
regions = np.array([
    "North",
    "South",
    "North",
    "East",
    "South",
    "North"
])

values, counts = np.unique(regions, return_counts=True)

print(values)
print(counts)
```

# 4. np.sort()

Sort an array:

```python
sales = np.array([400, 100, 250, 80, 300])

print(np.sort(sales))

print("\nFor decending order")
print(np.sort(sales)[:: -1]) # [::-1] means reverse the array
```

# 5. np.argsort() — 

This one is especially useful for Data Analytics.

np.sort() gives you the sorted values.

np.argsort() gives you the positions/indexes that would sort the values.

Example:

```python
sales = np.array([400, 100, 250, 80, 300])

print(np.argsort(sales))
```

# Output:

[3 1 2 4 0]

Why?

Original:

Index:   0    1    2   3    4
Sales:  400  100  250  80  300

Sorted:

80 → index 3
100 → index 1
250 → index 2
300 → index 4
400 → index 0

Therefore:

[3, 1, 2, 4, 0]

Use argsort() to sort another array

This is where it becomes powerful.

```python
import numpy as np
names = np.array([
    "Alice",
    "Bob",
    "Charlie",
    "David",
    "Emma"
])

sales = np.array([
    400,
    100,
    250,
    80,
    300
])

order = np.argsort(sales)

print(names[order])
print(sales[order])
```

# Mini Data Analysis Challenge

```python
import numpy as np

products = np.array([
    "Laptop",
    "Phone",
    "Tablet",
    "Monitor",
    "Keyboard",
    "Mouse",
    "Headphones",
    "Webcam"
])

sales = np.array([
    120,
    350,
    80,
    220,
    150,
    60,
    280,
    95
])

categories = np.array([
    "Computer",
    "Mobile",
    "Computer",
    "Computer",
    "Accessory",
    "Accessory",
    "Accessory",
    "Accessory"
])

print("Task 1")
performance = np.where(sales >= 150, "Good", "Low")
print("Product Performances", performance)

print("\nTask 2")
conditions = [
    sales < 100,
    (sales >= 100) & (sales <= 250),
    sales > 250
]
level = [
    "Low",
    "Medium",
    "High"
]
sales_level = np.select(conditions, level, default= "unknown")
print(sales_level)

print("\nTask 3")
values, counts = np.unique(categories, return_counts = True )
print("unique_categories=", values)
print("category_counts=", counts)

print("\nTask 4")
order = np.argsort(sales)
print(products[order])
print(sales[order])



print("\nTask 5")
order = np.argsort(sales[:: -1][:3])
print(products[order])
print(sales[order])
```
