```python
import numpy as np

sales = np.array([
    120,
    150,
    170,
    200,
    180,
    210,
    190

])

print("Sales:", sales)
print("Total Sales =", np.sum(sales))
print("Total Sales = ", np.mean(sales))
print("Total Sales = ",np.min(sales))
print("Standard Deviation =", np.std(sales))
print("Sales above 180 =", sales[sales > 180])
```

🏆 Exercise 1

Create:

temperature = np.array([30,32,31,29,28,35,36])

Print:

Average temperature
Highest temperature
Lowest temperature
Total temperature
Temperatures above 31

```python
import numpy as np
temperature = np.array([30,32,31,29,28,35,36])
print("Average temperature = ", np.mean(temperature))
print("Highest temperature = ", np.max(temperature))
print("Lowest temperature = ", np.min(temperature))
print("Total temperature = ", np.sum(temperature))
print("Temperatures above 31 = ", temperature[temperature > 31])
```

import numpy as np

ages = np.array([18, 22, 30, 25, 40, 35, 29, 21])

Print:

Mean age
Maximum age
Minimum age
Ages above 25
Ages below 25

```python
import numpy as np
ages = np.array([18, 22, 30, 25, 40, 35, 29, 21])

print("Mean age =",np.mean(ages))
print("Maximum age =",np.max(ages))
print("Minimum age =",np.min(ages))
print("Ages above 25 = ",ages [ages > 25])
print("Ages below 25 =", ages[ages < 25])
```

marks = np.array([

Without using loops, print:

Highest mark

Average mark
Sum of all marks
All marks greater than 85

This is your first exercise using a 2D NumPy array, which is much closer to working with real datasets.

```python
import numpy as np

marks = np.array([
    [75, 82, 91],
    [65, 88, 72],
    [95, 90, 85]
])

print("Highest mark =", np.max(marks))
print("Lowest mark =", np.min(marks))
print("Average mark =", np.mean(marks))
print("Sum of all marks =", np.sum(marks))
print("Marks greater than 85 are", marks[marks > 85])
```
