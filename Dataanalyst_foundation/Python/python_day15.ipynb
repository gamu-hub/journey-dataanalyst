```python
def count_even(numbers):
    count = 0
    for num in numbers:
        if num % 2 == 0:
            count += 1
    return count
print(count_even([1,2,3,4,5,6]))
```

```python
def count_odd (numbers):
    count = 0
    for num in numbers:
        if num %2 != 0:
            count += 1
    return count
print(count_odd([1,2,3,4,5,6,7]))
```

```python
def filter_even(numbers):
    result = []
    for num in numbers:
        if num % 2 == 0:
            result.append(num)
    return result
print(filter_even([1,2,3,4,5,6]))
```
