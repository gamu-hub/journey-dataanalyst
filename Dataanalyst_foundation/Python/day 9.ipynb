1. Dictionaries

A dictionary stores data in key : value format.

Example:

student = {
    "name": "John",
    "age": 20,
    "course": "Python"
}

print(student)

Output:

{'name': 'John', 'age': 20, 'course': 'Python'}

Think of it like:

KEY	VALUE
name	John
age	20
course	Python

```python
student = {
    "name": "John",
    "age" : 20,
    "course" : "Python"
}
print(student)
```
