```python
# Part 1 — What is a List?

# A list stores MANY values inside one variable.

# Example:

students = ["David", "Sarah", "Moshe"]
print(students)
```

```python
# Part 2 — Print a List

fruits = ["apple", "banana", "orange"]
print(fruits)
```

```python
# Part 3 — Access Items by Position

# Each item has an index number.

fruits = ["apple", "banana", "orange"]
print(fruits[2])
```

```python
# Exercise 1

# Create a list of 5 foods and print:

# the whole list
# the first food
# the last food

foods = ["pizza", "rice", "orange", "banana", "apple"]
print(foods[0])
print(foods[4])
```

```python
# Part 4 — Add Items to List


fruits = ["apple", "banana"]

fruits.append("orange")

print(fruits)
```

```python
# Exercise 2

# Create an empty list.

# Ask the user to enter 3 names.

# Add them using append().

# Print the final list.



names = []

name = input("Enter name: ")

names.append(name)
print(name)
```

```python
# Part 5 — Loop Through a List

# VERY important for data analytics.

numbers = [10, 20, 30]

for number in numbers:
    print(number)
```

```python
# Exercise 3

# Print each animal one by one.

animals = ["cat", "dog", "lion", "tiger"]
for animal in animals:
    print(animal)
```

```python
# Part 6 — Calculate Total Using Loop
numbers = [5, 10, 15]

total = 0

for number in numbers:
    total = total + number

print(total)
```

```python
# Exercise 4 (Important)

# Ask user to enter 5 numbers.

# Store them in a list.

# Then calculate:

# total
# average

numbers = []

for i in range(5):
    number = int(input("Enter a number: "))
    numbers.append(number)

total = 0

for number in numbers:
    total = total + number

average = total / 5

print("Numbers:", numbers)
print("Total:", total)
print("Average:", average)
```
