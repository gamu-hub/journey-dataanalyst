Why This Matters

When you become a Data Analyst, you will eventually write code like:

calculate_average(sales)

Instead of repeating code again and again, you create functions.

A reusable block of code that performs a task.

PART 2 — PARAMETERS

Functions become useful when they receive information.

Example:

def greet(name):

greet("John")

Hello John

PART 5 — FINDING VALUES

Problem:

Find largest number.

Think first.

Step 1

Start with first number.

largest = numbers[0]
Step 2

Loop through list.

for num in numbers:
Step 3

Compare.

if num > largest:
Step 4

Update.

largest = num
Full Function
def find_largest(numbers):
    largest = numbers[0]

    for num in numbers:
        if num > largest:
            largest = num

    return largest

PART 6 — FINDING SMALLEST

Same idea.

def find_smallest(numbers):
    smallest = numbers[0]

            smallest = num

print(find_smallest([7, 2, 9, 1, 5]))

1

PART 7 — COUNTING

A very common Data Analyst task.

Example:

Count even numbers.

def count_even(numbers):
    count = 0

    for num in numbers:
        if num % 2 == 0:
            count += 1

    return count

Example:

print(count_even([1,2,3,4,5,6]))

Output:

3

PART 8 — COUNT ODD NUMBERS
def count_odd(numbers):
    count = 0

    for num in numbers:
        if num % 2 != 0:
            count += 1

    return count

PART 9 — SEARCHING

Search for a value.

def find_number(numbers, target):
    for num in numbers:
        if num == target:
            return True

    return False

Example:

print(find_number([1,2,3,4], 3))

Output:

True

def find_name(names, target):

PART 11 — FILTERING

Filtering means:

Keep some items and remove others.

Example:

Keep numbers greater than 5.

def greater_than_five(numbers):
    result = []

    for num in numbers:
        if num > 5:
            result.append(num)

    return result

Output:

[8, 10, 7]

PART 13 — AVERAGE

Very important for data analysis.

Formula:

ext{Average}=rac{ ext{Sum of Numbers}}{ ext{Count of Numbers}}

Code:

def calculate_average(numbers):
    total = 0

    for num in numbers:
        total += num

    average = total / len(numbers)

    return average

Example:

print(
    calculate_average(
        [10,20,30]
    )
)

Output:

20

PART 14 — COUNT VOWELS
def count_vowels(word):
    count = 0

    vowels = "aeiou"

    for letter in word:
        if letter in vowels:
            count += 1

    return count

Example:

print(count_vowels("education"))

Output:

5

PART 15 — LONGEST WORD
def longest_word(words):
    longest = words[0]

    for word in words:
        if len(word) > len(longest):
            longest = word

    return longest

Example:

print(
    longest_word(
        ["cat", "elephant", "dog"]
    )
)

Output:

elephant
