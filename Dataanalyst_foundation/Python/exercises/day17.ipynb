Exercise 1 — Read a CSV File
Step 1

Create:

products.csv

Put inside:

product,price
Phone,500
Laptop,1200
Mouse,50

Task

Print all rows.

Expected output:

['product', 'price']
['Phone', '500']
['Laptop', '1200']
['Mouse', '50']

```python
#lets make a csv file first. 

import csv
with open("products.csv", "w", newline= "") as file:
    writer = csv.writer(file)

    writer.writerow(["product", "price"])
    writer.writerow(["Phone", 500 ])
    writer.writerow(["Laptop", 1200])
    writer.writerow(["Mouse", 50])
```

```python
import csv

with open("products.csv", "r") as file:
    reader = csv.reader(file)
    for row in reader:
        print(row)
```

Mini Challenge 🔥

Create:

students.csv
name,age,grade
John,20,A
Anna,21,B

Try printing ONLY:

ages
grades
then all together in one sentence 🚀

```python
import csv
with open("students.csv", "w", newline= "") as file:
    writer = csv.writer(file)
    writer.writerow(["name", "age", "grade"])
    writer.writerow(["John", 20, "A"])
    writer.writerow(["Anna", 21, "B"])

with open("students.csv", "r") as file:
    reader = csv.reader(file)
    next(reader)

    for row in reader:
        print(f"{row[0]} is {row[1]}")

```

```python
import csv
with open("products.csv", "r") as file:
    reader = csv.reader(file)
    next(reader)
      
    for row in reader:
        product = row[0]
        price = int(row[1])

        if price > 1000:
            print(f"{product} costs {price}")

    
```

Next Challenge — Count Expensive Products 🔥

```python
import csv
count = 0
with open ("products.csv", "r") as file:
    reader = csv.reader(file)
    next(reader)

    for row in reader:
        product = row[0]
        price = int(row[1])
        
        if price > 1000:
            count += 1
print(f"Total expensinve prouducts: {count}")

          
```
