Mini Practice for You
students = [
    {"name": " john ", "score": 90},
    {"name": "SARAH", "score": 85},
    {"name": " mike", "score": 70}
]

Try to:

Clean the names
Print them properly
Sort by score

```python
students = [
    {"name": "  john", "score": 90},
    {"name": "SARAH", "score": 85},
    {"name": "mike","score": 70 }
]

for student in students:
    student["name"] = student["name"].strip().title()
sorted_students = sorted(students, key = lambda x:x["score"])

for student in sorted_students:
    print(student["name"], student["score"])
```
