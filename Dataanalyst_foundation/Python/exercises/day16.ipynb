Exercise 1 — Print All Items


```python
numbers = [5, 10, 15, 20]

for number in numbers:
    print(number)
```

Exercise 2 — Find Total

```python
numbers = [1, 2, 3, 4, 5]

total = 0

for number in numbers:
    total += number

print(total)
```

Exercise 3 — Find Largest Number

```python
numbers = [4, 9, 2, 15, 7]

largest = numbers[0]

for number in numbers:
    if number > largest:
        largest = number

print(largest)
```

Exercise 4 — Count Even Numbers

```python
numbers = [1, 2, 4, 5, 8, 9]

count = 0

for number in numbers:
    if number % 2 == 0:
        count += 1

print(count)
```

Exercise 5 — Shopping List

```python
shopping_list = []

for i in range(3):
    item = input("Enter item: ")
    shopping_list.append(item)

print(shopping_list)
```

Exercise 6 — Search Name

```python
names = ["John", "Sarah", "David"]

search = input("Enter name: ")

if search in names:
    print("Found")
else:
    print("Not Found")
```

Exercise 7 — Remove Duplicate Numbers

```python
numbers = [1, 1, 2, 2, 3, 4, 4]

unique_numbers = []

for number in numbers:
    if number not in unique_numbers:
        unique_numbers.append(number)

print(unique_numbers)
```

Exercise 8 — Largest & Smallest Together

```python
numbers = [5, 8, 1, 20, 3]

largest = numbers[0]
smallest = numbers[0]

for number in numbers:
    if number > largest:
        largest = number

    if number < smallest:
        smallest = number

print("Largest:", largest)
print("Smallest:", smallest)
```

BONUS CHALLENGE — Number Analyzer 🔥

```python
numbers = []

for i in range(5):
    number = int(input("Enter a number: "))
    numbers.append(number)

print("Numbers:", numbers)

total = 0

for number in numbers:
    total += number

print("Total:", total)

average = total / len(numbers)

print("Average:", average)

largest = numbers[0]

for number in numbers:
    if number > largest:
        largest = number

print("Largest:", largest)

smallest = numbers[0]

for number in numbers:
    if number < smallest:
        smallest = number

print("Smallest:", smallest)

print("Even numbers:")

for number in numbers:
    if number % 2 == 0:
        print(number)
```
