Challenge 4:

numbers = [8, 3, 15, 1, 9]

Find the smallest number.

```python
numbers = [8,3,15,1,9]
smallest = numbers[0]
for num in numbers:
    if num < smallest:
        smallest = num
print(smallest)
```
