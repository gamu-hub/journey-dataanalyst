```python
# Exercise 4 (Important)

# Ask user to enter 5 numbers.

# Store them in a list.

# Then calculate:

# total
# average

numbers = []

for i in range(5):
    number = int(input("Enter a number: "))
    numbers.append(number)

total = 0

for number in numbers:
    total = total + number

average = total / 5

print("Numbers:", numbers)
print("Total:", total)
print("Average:", average)
```
