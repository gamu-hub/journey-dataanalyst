Exercise 6 — functions with logic.

Create a function called find_biggest().

Rules:

takes 2 numbers
returns the bigger number

Example:

print(find_biggest(5, 9))

```python
def find_biggest(a,b):
    if a > b:
        return a
    else:
        return b
print(find_biggest(4,7))
print(find_biggest(3,10))
```

Rules:
