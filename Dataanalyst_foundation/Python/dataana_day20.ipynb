```python
def square(x):
    return x * x
```

```python
square = lambda x: x * x
print(square(5))
```

```python
numbers = [1, 2, 3, 4]
result = map(lambda x: x * x, numbers)
print(list(result))
```

```python
numbers = [1, 2, 3, 4, 5, 6]

result = filter(lambda x: x % 2 == 0, numbers)

print(list(result))
```

```python
people = [
    {"name": "John", "age": 25},
    {"name": "Sarah", "age": 20},
    {"name": "Mike", "age": 30}

]

sorted_people = sorted(people, key = lambda x: x["age"])
print(sorted_people)
```
