```python
import pandas as pd

data = {
    "Team": [
        "Team HOPE", "ELV esports", "PLE esports",
        "Team HOPE", "ELV esports", "PLE esports"
    ],

    "Country": [
        "Israel", "USA", "Korea",
        "Israel", "USA", "Korea"
    ],
    "Kills": [40, 35, 50, 45, 30, 55],
    "Matches" : [ 5, 5, 5, 5, 5, 5]
}

df = pd.DataFrame(data)
average_kills = df.groupby("Team")["Kills"].mean()

print(average_kills)
```

```python
import pandas as pd

data = {
    "Team": [
        "Team HOPE", "ELV esports", "PLE esports",
        "Team HOPE", "ELV esports", "PLE esports"
    ],

    "Country": [
        "Israel", "USA", "Korea",
        "Israel", "USA", "Korea"
    ],
    "Kills": [40, 35, 50, 45, 30, 55],
    "Matches" : [ 5, 5, 5, 5, 5, 5]
}

df = pd.DataFrame(data)
stats = df.groupby("Team")["Kills"].agg(["sum", "mean", "max", "min"])
print(stats)
```

```python
import pandas as pd

data = {
    "Teams": [
        "Team HOPE",
        "SOUL esports",
        "MVL esports",
        "Team HOPE",
        "SOUL esports",
        "MVL esports"
        
    ],
    "Kills" : [ 40, 35, 50, 45, 30, 55],
    "Matches" : [5, 5, 5, 5, 5, 5]
}

df = pd.DataFrame(data)
result = df.groupby("Teams")["Kills"].agg(["sum", "mean", "max", "min"])

print(result)
```
