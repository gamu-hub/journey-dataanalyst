6. Multiple Conditions (&)
Teams with kills above 30 AND from Korea
result = df[(df["Kills"] > 30) & (df["Country"] == "Korea")]

print(result)

```python
import pandas as pd

data = {
    "Team": ["Team HOPE", "ELV esports", "Muvanlai esports", "PLE esports"],
    "Kills": [40, 35, 28, 50],
    "Matches": [5, 5, 5, 5],
    "Country": ["Israel", "USA", "India", "Korea"]
}

df = pd.DataFrame(data)

result = df[(df["Kills"] > 30) & (df["Country"] == "Korea")]
print(result)
```

7. OR Condition (|)
Teams from Israel OR India

```python
import pandas as pd

data = {
    "Team": ["Team HOPE", "ELV esports", "Muvanlai esports", "PLE esports"],
    "Kills": [40, 35, 28, 50],
    "Matches": [5, 5, 5, 5],
    "Country": ["Israel", "USA", "India", "Korea"]
}

df = pd.DataFrame(data)

result = df[(df["Country"] == "Israel") | (df["Country"] == "India")]
print(result)
```
