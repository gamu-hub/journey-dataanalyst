2. Updating an Entire Column

You can replace all values in a column.

```python
import pandas as pd

data = {
    "Team": ["Team HOPE", "ELV esports", "Muvanlai esports", "PLE esports"],
    "Kills": [40, 35, 28, 50],
    "Matches": [5, 5, 5, 5],
    "Country": ["Israel", "USA", "India", "Korea"]
}

df = pd.DataFrame(data)
df["Matches"] = 10
print(df)
```

3. Updating Specific Rows with .loc[]

.loc[] is one of the MOST important Pandas tools.

Syntax:

df.loc[row, column] = value

```python
import pandas as pd

data = {
    "Team": ["Team HOPE", "ELV esports", "Muvanlai esports", "PLE esports"],
    "Kills": [40, 35, 28, 50],
    "Matches": [5, 5, 5, 5],
    "Country": ["Israel", "USA", "India", "Korea"]
}

df = pd.DataFrame(data)

df.loc[1, "Kills"] = 45
print(df)
```

4. Updating Multiple Columns in One Row

```python
import pandas as pd

data = {
    "Team": ["Team HOPE", "ELV esports", "Muvanlai esports", "PLE esports"],
    "Kills": [40, 35,28,50],
    "Matches": [5,5,5,5],
    "Country": ["Israel", "USA", "India", "Korea"]
}

df = pd.DataFrame(data)
df.loc[2, ["Kills", "Country"]] = [30, "Japan"]
print(df)
```

5. Conditional Updates (Very Important)

This is real data-cleaning style.

Example:

Increase kills for teams with less than 40 kills.

df.loc[df["Kills"] < 40, "Kills"] = 40

print(df)

```python
import pandas as pd

data = {
    "Team": ["Team HOPE", "ELV esports", "Muvanlai esports", "PLE esports"],
    "Kills": [40, 35,28,50],
    "Matches": [5,5,5,5],
    "Country": ["Israel", "USA", "India", "Korea"]
}

df = pd.DataFrame(data)
df.loc[df["Kills"] < 40, "Kills"] = 40
print(df)
```
