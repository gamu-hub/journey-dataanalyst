4. Different Types of Merge
A. Inner Merge (Default)

Only matching rows appear.

pd.merge(df1, df2, on="ID", how="inner")

```python
import pandas as pd

df1 = pd.DataFrame({
    "ID": [1, 2, 3],
    "Name": ["A", "B", "C"]
})

df2 = pd.DataFrame({
    "ID": [2, 3, 4],
    "Score": [80, 90, 70]
})

result = pd.merge(df1, df2, on="ID", how="inner")

print(result)
```

B. Left Merge

Keeps ALL rows from left dataframe.

pd.merge(df1, df2, on="ID", how="left")

```python
import pandas as pd

df1 = pd.DataFrame({
    "ID": [1, 2, 3],
    "Name": ["A", "B", "C"]
})

df2 = pd.DataFrame({
    "ID": [2, 3, 4],
    "Score": [80, 90, 70]
})

pd.merge(df1, df2, on="ID", how="left")

print(result)
```

DIFFERENCE BETWEEN ALL TYPES OF MERGE

```python
import pandas as pd

df1 = pd.DataFrame({
    "ID": [1, 2, 3],
    "Name": ["Alex", "John", "Mika"]
})

df2 = pd.DataFrame({
    "ID": [2, 3, 4],
    "Score": [80, 90, 70]
})

print("INNER")
print(pd.merge(df1, df2, on="ID", how="inner"))

print("\nLEFT")
print(pd.merge(df1, df2, on="ID", how="left"))

print("\nRIGHT")
print(pd.merge(df1, df2, on="ID", how="right"))

print("\nOUTER")
print(pd.merge(df1, df2, on="ID", how="outer"))
```

6. Merge Using Different Column Names

Sometimes column names are different.

```python
df1 = pd.DataFrame({
    "TeamID": [1, 2],
    "Team": ["HOPE", "ELV"]
})

df2 = pd.DataFrame({
    "ID": [1, 2],
    "Kills": [40, 35]
})

result = pd.merge(
    df1,
    df2,
    left_on="TeamID",
    right_on="ID"
)

print(result)
```

```python
df1 = pd.DataFrame({
    "TeamID": [1, 2],
    "Team": ["HOPE", "ELV"]
})

df2 = pd.DataFrame({
    "ID": [1, 2],
    "Kills": [40, 35]
})

result = pd.merge(
    df1,
    df2,
    left_on="TeamID",
    right_on= "ID"
)

print(result)
```

7. Real Data Analyst Example

```python
import pandas as pd

teams = pd.DataFrame({
    "PlayerID": [101, 102, 103],
    "Player": ["Alex", "John", "Mika"]
})

stats = pd.DataFrame({
    "PlayerID": [101, 102, 103],
    "Kills": [15, 20, 18],
    "Matches": [5, 5, 5]
})

result = pd.merge(teams, stats, on="PlayerID")

result["KD"] = result["Kills"] / result["Matches"]

print(result)
```
