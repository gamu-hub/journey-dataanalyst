Why is this useful for a Data Analyst?

Suppose you're analyzing customer spending:

Q1 = ₪100
Q2 = ₪180
Q3 = ₪350

You can understand the distribution much better than by looking only at the average.

For example:

25% of customers spend around/below ₪100
50% are around/below ₪180
75% are around/below ₪350

This is particularly useful when the data isn't evenly distributed.

```python
import numpy as np

sales = np.array([120, 150, 150, 180, 200, 220, 150])

q1 = np.percentile(sales, 25)
q2 = np.percentile(sales, 50)
q3 = np.percentile(sales, 75)

print("Q1:", q1)
print("Q2:", q2)
print("Q3:", q3)
```

2️⃣ How do we detect outliers?

This is where your IQR becomes useful.

Remember:

IQR = Q3 - Q1

We calculate two boundaries:

Lower boundary
Q1 - 1.5 × IQR
Upper boundary
Q3 + 1.5 × IQR

```python
# Let's do it with Python
import numpy as np

sales = np.array([50, 55, 60, 62, 65, 70, 72, 75, 80, 500])

q1 = np.percentile(sales, 25)
q3 = np.percentile(sales, 75)

iqr = q3 - q1

lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

print("Q1:", q1)
print("Q3:", q3)
print("IQR:", iqr)
print("Lower bound:", lower_bound)
print("Upper bound:", upper_bound)

outliers = sales[(sales < lower_bound) | (sales > upper_bound)]

print("Outliers:", outliers)
```

# Lesson 6 — Z-Score

You already learned standard deviation. Now we're going to use it to answer a more specific question:

How far away is one value from the average?

```python
import numpy as np

sales = np.array([100, 110, 120, 130, 140])

mean = np.mean(sales)
std = np.std(sales)

value = 140

z_score = (value - mean) / std

print("Mean:", mean)
print("Standard deviation:", std)
print("Z-score:", z_score)
```
