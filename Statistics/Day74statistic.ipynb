1️⃣ What is Probability?

Probability measures how likely something is to happen.

It ranges from:

0 ─────────────────────── 1
Impossible              Certain

Or as percentages:

0% ───────────────────── 100%

5️⃣ Probability from Data

You don't always need a theoretical experiment.

As a Data Analyst, you'll often calculate probability from observed data.

Suppose:

orders = 1000
returned = 75

Then:

probability_return = returned / orders

print(probability_return)

Output:

0.075

That's:

7.5%

You could tell your manager:

"Based on the historical data, approximately 7.5% of orders were returned."

```python
# exercises
total_customers = 1000
premium_customers = 250

print("Task 1")
probability = premium_customers / total_customers
print(probability)

print("\nTask 2")
percentage = probability * 100
print(percentage)

print("\nTask 3")
print("There is a 25% probability that a randomly selected customer is a premium customer")
```

# Probability Lesson 2: Complementary Probability

What is a complement?
The complement means:
The probability that something does NOT happen.

If:

P(Event) = 25%

then:

P(Not Event) = 75%

Because all possibilities together must equal 100%.

Formula
P(Not A) = 1 - P(A)

# FOR EXAMPLE

P(Premium) = 0.25

P(Not Premium) = 1 - 0.25
               = 0.75

```python
# EXERCISE
total_orders = 5000
failed_orders = 200

print("Task 1")
failure_probability = failed_orders / total_orders
print(failure_probability)

print("\nTask 2")
success_probability = 1 - failure_probability
print(success_probability)

print("\nTask 3")
percentage = success_probability * 100
print(percentage)

print("\nTask 4")
print("Based on the data, approximately 96% of orders are successful")

print("\none suggestion for task 3")

failure_percentage = failure_probability * 100
success_percentage = success_probability * 100
print("Failure:", failure_percentage, "%")
print("Success:", success_percentage, "%")

```

# Independent Events

Two events are independent when one event does not affect the other.

Example: Coin tosses 🪙

You toss a coin:

Heads / Tails

You get:

Heads

Now you toss it again.

Does the first toss change the second toss?

No.

The second toss still has:

50% Heads
50% Tails

So the two tosses are independent.

# Dependent Events

Two events are dependent when the first event changes the probability of the second event.

Example: Drawing cards 🃏

Imagine you have:

10 cards

and:

3 are red
7 are blue

You draw one card without putting it back.

If you draw a red card:

Before:
3 red / 10 total

After drawing it:

2 red / 9 total

The probability changed.

Therefore, the events are dependent.

# Easy way to remember

Ask yourself:

"Does the first event change the second event?"

If:

NO → Independent
YES → Dependent

```python
# Exercises

print("Task 1 = Independent")
print("Task 2 = Dependent")
print("Task 3 = Dependent because the customer bought another product because he got the discount coupon upon buying the first product")
print("Task 4 = Independent because the second customer is completely unrelated ")
```

Lesson 4: Combined Probability

Now we're going to calculate probabilities involving two events.

For independent events, we multiply their probabilities:

$$ P(A\text{ and }B) = P(A) \times P(B) $$
Simple example

You flip a coin twice.

Probability of Heads:

1 / 2 = 0.5

Probability of getting Heads AND Heads:

0.5 × 0.5 = 0.25

So:

25%

The reason we can multiply is because the two coin flips are independent.

# Your first combined-probability exercise

A company has:

total_customers = 1000
premium_customers = 250

So:

P(Premium) = 0.25

Now imagine that 20% of premium customers make a second purchase.

P(Second purchase | Premium) = 0.20

For now, we'll treat this as a multiplication exercise.

We have:

25% of all customers are premium.
20% of premium customers make a second purchase.

Imagine 100 customers.

100 customers
     ↓
25 are premium
     ↓
20% of those 25 make a second purchase
     ↓
5 customers

So the question:

"What percentage of ALL customers are both premium AND make a second purchase?"

is asking:
25 out of 100
AND
20% of those 25

That's why we multiply:

25% × 20%

or as probabilities:

0.25 × 0.20

```python
# EXERCISE
# We have 200 customers in total
total_customers = 200

# 50 of those customers are Premium customers
premium_customers = 50

# 10 Premium customers make a second purchase
second_purchase_premium = 10


# --------------------------------
# Task 1
# What percentage of ALL customers are Premium?
# --------------------------------

# Probability = Premium customers / Total customers
premium_probability = premium_customers / total_customers

print("Task 1")
print("Premium probability:", premium_probability)

# Convert the probability into a percentage
premium_percentage = premium_probability * 100

print("Premium percentage:", premium_percentage, "%")


# --------------------------------
# Task 2
# What percentage of Premium customers
# make a second purchase?
# --------------------------------

# Here our total group is ONLY Premium customers.
# We already know these customers are Premium.
second_purchase_probability = second_purchase_premium / premium_customers

print("\nTask 2")
print("Second purchase probability:", second_purchase_probability)

# Convert to percentage
second_purchase_percentage = second_purchase_probability * 100

print("Second purchase percentage:", second_purchase_percentage, "%")


# --------------------------------
# Task 3
# What percentage of ALL customers are BOTH:
# Premium AND make a second purchase?
# --------------------------------

# We have 10 customers who satisfy BOTH conditions.
# Our total group is ALL 200 customers.
combined_probability = second_purchase_premium / total_customers

print("\nTask 3")
print("Combined probability:", combined_probability)

# Convert to percentage
combined_percentage = combined_probability * 100

print("Combined percentage:", combined_percentage, "%")
```
