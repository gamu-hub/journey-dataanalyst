# Suppose a company runs a promotion:

Outcome	Probability
Customer spends ₪100	30%
Customer spends ₪50	70%

Convert percentages to decimals:

0.30
0.70

Now calculate:

expected_value = (100 * 0.30) + (50 * 0.70)

Result:

65.0

```python
outcome_1 = 100
probability_1 = 0.30

outcome_2 = 50
probability_2 = 0.70

expected_value = (
    outcome_1 * probability_1
    + outcome_2 * probability_2
)

print("Expected Value:", expected_value)
```
